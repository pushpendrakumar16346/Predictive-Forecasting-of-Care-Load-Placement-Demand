````python
from pathlib import Path
import os
import json
import csv
import hashlib
import shutil
import zipfile
import ast
import pandas as pd
from datetime import datetime

PROJECT_ROOT = Path(
    r"C:\Users\hp\Predictive Forecasting of Care Load & Placement Demand"
)

QA_DIR = PROJECT_ROOT / "outputs" / "final_quality_assurance"
INVENTORY_FILE = QA_DIR / "final_project_inventory.csv"

QA_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 75)
print("FINAL PROJECT INVENTORY AND VERIFICATION")
print("=" * 75)

# Final project inventory
excluded = {
    ".deployment_test_env",
    ".deployment_test_env",
    "__pycache__",
    ".git"
}

inventory = []

for path in PROJECT_ROOT.rglob("*"):
    if not path.is_file():
        continue

    if any(part in excluded for part in path.parts):
        continue

    try:
        inventory.append({
            "file": str(path.relative_to(PROJECT_ROOT)),
            "extension": path.suffix.lower(),
            "size_bytes": path.stat().st_size
        })
    except OSError:
        pass

inventory_df = pd.DataFrame(inventory)
inventory_df.to_csv(INVENTORY_FILE, index=False)

print("\nPROJECT INVENTORY")
print("Files inventoried:", len(inventory_df))
print("Inventory saved:", INVENTORY_FILE)

# Documentation verification
documentation_files = [
    "README.md",
    "INSTALLATION_GUIDE.md",
    "USER_GUIDE.md",
    "TECHNICAL_DOCUMENTATION.md"
]

documentation_results = []

for filename in documentation_files:
    path = PROJECT_ROOT / filename
    documentation_results.append({
        "file": filename,
        "exists": path.exists(),
        "non_empty": path.exists() and path.stat().st_size > 0
    })

documentation_df = pd.DataFrame(documentation_results)
documentation_df.to_csv(
    QA_DIR / "documentation_verification.csv",
    index=False
)

print("\nDOCUMENTATION VERIFICATION")
print(documentation_df.to_string(index=False))

# Original requirement evidence
requirements = {
    "Data loading and inspection": ["read_csv", "read_excel"],
    "Data cleaning": ["dropna", "to_datetime"],
    "Date continuity": ["date_range", "missing_dates"],
    "Exploratory analysis": ["describe", "correlation"],
    "Trend and seasonality": ["seasonal_decompose", "trend"],
    "Lag and rolling features": ["shift(", "rolling("],
    "Flow indicators": ["transfers", "discharges"],
    "Calendar features": ["day_of_week", "day_of_year"],
    "Chronological validation": ["TimeSeriesSplit", "chronological"],
    "Baseline forecasting": ["persistence", "moving average"],
    "Statistical forecasting": ["ARIMA", "ExponentialSmoothing"],
    "Machine learning forecasting": ["RandomForest", "GradientBoosting"],
    "Multiple forecast horizons": ["1, 7, 14, 30", "FORECAST_HORIZONS"],
    "Forecast evaluation": ["MAE", "RMSE", "MAPE"],
    "Prediction uncertainty": ["confidence_interval", "lower_bound"],
    "Capacity and surge risk": ["capacity", "surge"],
    "Scenario analysis": ["scenario"],
    "Dashboard": ["streamlit", "st.dataframe"]
}

source_extensions = {".py", ".ipynb", ".md"}
source_files = [
    p for p in PROJECT_ROOT.rglob("*")
    if p.is_file()
    and p.suffix.lower() in source_extensions
    and not any(part in excluded for part in p.parts)
]

source_text = ""

for path in source_files:
    try:
        source_text += "\n" + path.read_text(
            encoding="utf-8",
            errors="ignore"
        )
    except OSError:
        pass

requirement_results = []

for requirement, keywords in requirements.items():
    matches = [
        keyword for keyword in keywords
        if keyword.lower() in source_text.lower()
    ]

    requirement_results.append({
        "requirement": requirement,
        "keyword_evidence": len(matches),
        "evidence_found": "; ".join(matches),
        "status": "EVIDENCE FOUND" if matches else "REVIEW REQUIRED"
    })

requirements_df = pd.DataFrame(requirement_results)
requirements_df.to_csv(
    QA_DIR / "original_requirements_verification.csv",
    index=False
)

print("\nORIGINAL REQUIREMENTS EVIDENCE")
print(requirements_df.to_string(index=False))

# Notebook inventory
notebook_files = list((PROJECT_ROOT / "notebooks").glob("*.ipynb"))
notebook_results = []

for notebook in notebook_files:
    try:
        with open(notebook, "r", encoding="utf-8") as f:
            nb = json.load(f)

        code_cells = [
            cell for cell in nb.get("cells", [])
            if cell.get("cell_type") == "code"
        ]

        syntax_errors = []

        for index, cell in enumerate(code_cells):
            code = "".join(cell.get("source", []))
            if not code.strip():
                continue

            try:
                ast.parse(code)
            except SyntaxError as error:
                syntax_errors.append(
                    f"Cell {index}: {error.msg}"
                )

        notebook_results.append({
            "notebook": notebook.name,
            "code_cells": len(code_cells),
            "syntax_errors": len(syntax_errors),
            "syntax_status": "PASS" if not syntax_errors else "FAIL",
            "details": "; ".join(syntax_errors)
        })

    except Exception as error:
        notebook_results.append({
            "notebook": notebook.name,
            "code_cells": 0,
            "syntax_errors": 1,
            "syntax_status": "FAIL",
            "details": str(error)
        })

notebook_df = pd.DataFrame(notebook_results)
notebook_df.to_csv(
    QA_DIR / "notebook_syntax_verification.csv",
    index=False
)

print("\nNOTEBOOK SYNTAX VERIFICATION")
print(notebook_df.to_string(index=False))

# Model and output verification
model_directories = [
    PROJECT_ROOT / "models",
    PROJECT_ROOT / "outputs"
]

model_results = []

for directory in model_directories:
    if not directory.exists():
        continue

    for path in directory.rglob("*"):
        if not path.is_file():
            continue

        if any(part in excluded for part in path.parts):
            continue

        if path.suffix.lower() not in {
            ".joblib", ".pkl", ".json", ".csv"
        }:
            continue

        status = "PASS"
        details = "File exists and is readable"

        try:
            if path.suffix.lower() == ".json":
                json.loads(path.read_text(encoding="utf-8"))

            elif path.suffix.lower() == ".csv":
                pd.read_csv(path, nrows=5)

            elif path.suffix.lower() in {".joblib", ".pkl"}:
                if path.stat().st_size == 0:
                    raise ValueError("Empty model artifact")

        except Exception as error:
            status = "FAIL"
            details = str(error)

        model_results.append({
            "file": str(path.relative_to(PROJECT_ROOT)),
            "status": status,
            "details": details
        })

model_df = pd.DataFrame(model_results)
model_df.to_csv(
    QA_DIR / "model_and_output_verification.csv",
    index=False
)

print("\nMODEL AND OUTPUT VERIFICATION")
print("Files checked:", len(model_df))
if not model_df.empty:
    print(model_df["status"].value_counts().to_string())

# Dashboard verification
app_path = PROJECT_ROOT / "app.py"
dashboard_status = "FAIL"
dashboard_details = "app.py not found"

if app_path.exists():
    try:
        app_code = app_path.read_text(
            encoding="utf-8",
            errors="ignore"
        )
        ast.parse(app_code)

        dashboard_status = "PASS"
        dashboard_details = "Dashboard source syntax valid"

    except Exception as error:
        dashboard_details = str(error)

dashboard_result = pd.DataFrame([{
    "file": "app.py",
    "status": dashboard_status,
    "details": dashboard_details
}])

dashboard_result.to_csv(
    QA_DIR / "dashboard_verification.csv",
    index=False
)

print("\nDASHBOARD VERIFICATION")
print(dashboard_result.to_string(index=False))

# Research and executive summary verification
deliverables = [
    PROJECT_ROOT / "research_paper" /
    "Predictive_Forecasting_Research_Paper.docx",
    PROJECT_ROOT / "research_paper" /
    "Predictive_Forecasting_Research_Paper.pdf",
    PROJECT_ROOT / "government_executive_summary" /
    "Government_Executive_Summary.docx",
    PROJECT_ROOT / "government_executive_summary" /
    "Government_Executive_Summary.pdf"
]

deliverable_results = []

for path in deliverables:
    deliverable_results.append({
        "file": str(path.relative_to(PROJECT_ROOT)),
        "exists": path.exists(),
        "size_bytes": path.stat().st_size if path.exists() else 0,
        "status": (
            "PASS" if path.exists() and path.stat().st_size > 0
            else "FAIL"
        )
    })

deliverable_df = pd.DataFrame(deliverable_results)
deliverable_df.to_csv(
    QA_DIR / "research_and_executive_summary_verification.csv",
    index=False
)

print("\nRESEARCH AND EXECUTIVE SUMMARY")
print(deliverable_df.to_string(index=False))

# Final verification summary
summary = {
    "project_files_inventoried": len(inventory_df),
    "documentation_files_passed": int(
        documentation_df["non_empty"].sum()
    ),
    "documentation_files_total": len(documentation_df),
    "requirements_with_keyword_evidence": int(
        (requirements_df["status"] == "EVIDENCE FOUND").sum()
    ),
    "requirements_total": len(requirements_df),
    "notebooks_found": len(notebook_df),
    "notebooks_with_valid_syntax": int(
        (notebook_df["syntax_status"] == "PASS").sum()
    ),
    "model_and_output_files_checked": len(model_df),
    "model_and_output_failures": (
        int((model_df["status"] == "FAIL").sum())
        if not model_df.empty else 0
    ),
    "dashboard_source_status": dashboard_status,
    "research_and_summary_files_passed": int(
        (deliverable_df["status"] == "PASS").sum()
    ),
    "research_and_summary_files_total": len(deliverable_df)
}

summary_path = QA_DIR / "final_quality_assurance_summary.json"
summary_path.write_text(
    json.dumps(summary, indent=4),
    encoding="utf-8"
)

print("\n" + "=" * 75)
print("FINAL QUALITY ASSURANCE SUMMARY")
print("=" * 75)

for key, value in summary.items():
    print(f"{key}: {value}")

print("\nSUMMARY SAVED:", summary_path)
print("\nQUALITY ASSURANCE INVENTORY AND CHECKS COMPLETED")
````

````python
from pathlib import Path
from datetime import datetime
import shutil
import hashlib
import zipfile
import pandas as pd
import json

from pptx import Presentation
from pptx.util import Inches, Pt

PROJECT_ROOT = Path(
    r"C:\Users\hp\Predictive Forecasting of Care Load & Placement Demand"
)

print("=" * 75)
print("FINAL PRESENTATION AND SUBMISSION PREPARATION")
print("=" * 75)

# Create final presentation
presentation_dir = PROJECT_ROOT / "final_presentation"
presentation_dir.mkdir(parents=True, exist_ok=True)

presentation_path = (
    presentation_dir / "Predictive_Forecasting_Final_Presentation.pptx"
)

prs = Presentation()

def add_slide(title, bullets):
    slide = prs.slides.add_slide(prs.slide_layouts[1])
    slide.shapes.title.text = title

    text_frame = slide.placeholders[1].text_frame
    text_frame.clear()

    for index, bullet in enumerate(bullets):
        paragraph = (
            text_frame.paragraphs[0]
            if index == 0
            else text_frame.add_paragraph()
        )
        paragraph.text = bullet
        paragraph.level = 0
        paragraph.font.size = Pt(21)

slide = prs.slides.add_slide(prs.slide_layouts[0])
slide.shapes.title.text = "Predictive Forecasting of Care Load"
slide.placeholders[1].text = (
    "Care Load and Placement Demand Forecasting\n"
    "Final Project Presentation"
)

add_slide("Project Overview", [
    "Time-series forecasting of children in HHS care",
    "Historical dataset containing 720 observations",
    "Observation period: January 2023 to December 2025",
    "Forecast horizons: 1, 7, 14 and 30 days"
])

add_slide("Data Preparation", [
    "Raw and processed dataset validation",
    "Date parsing and chronological organization",
    "Missing-date and data-quality checks",
    "Lag, rolling and calendar-based features"
])

add_slide("Forecasting Methodology", [
    "Persistence and moving-average baselines",
    "ARIMA and Exponential Smoothing",
    "Random Forest and Gradient Boosting",
    "Chronological validation and walk-forward evaluation"
])

add_slide("Forecast Evaluation", [
    "Mean Absolute Error (MAE)",
    "Root Mean Squared Error (RMSE)",
    "Mean Absolute Percentage Error (MAPE)",
    "Horizon-wise model evaluation and uncertainty analysis"
])

add_slide("Dashboard Modules", [
    "Historical monitoring",
    "Predictive forecasting",
    "Capacity and surge-risk analysis",
    "Scenario analysis and model comparison"
])

add_slide("Deployment and Monitoring", [
    "Separate deployment environment",
    "Model loading and registry",
    "Reproducibility and data-refresh checks",
    "Monitoring, drift and retraining documentation"
])

add_slide("Project Deliverables", [
    "Research paper in DOCX and PDF formats",
    "Government executive summary in DOCX and PDF formats",
    "README and installation, user and technical guides",
    "Streamlit dashboard and trained model artifacts"
])

add_slide("Conclusion", [
    "Integrated forecasting and dashboard framework",
    "Multiple forecast horizons and model families",
    "Deployment and maintenance documentation",
    "Final validation evidence retained with project outputs"
])

prs.save(presentation_path)

print("\nPRESENTATION CREATED:", presentation_path)

# Prepare final submission folder
submission_dir = PROJECT_ROOT / "final_submission"

if submission_dir.exists():
    shutil.rmtree(submission_dir)

submission_dir.mkdir(parents=True)

folders_to_copy = [
    "notebooks",
    "src",
    "dashboard",
    "models",
    "config",
    "documentation",
    "research_paper",
    "government_executive_summary"
]

files_to_copy = [
    "app.py",
    "README.md",
    "INSTALLATION_GUIDE.md",
    "USER_GUIDE.md",
    "TECHNICAL_DOCUMENTATION.md",
    "requirements.txt",
    "requirements-deployment.txt"
]

for folder_name in folders_to_copy:
    source = PROJECT_ROOT / folder_name

    if source.exists() and source.is_dir():
        shutil.copytree(
            source,
            submission_dir / folder_name,
            ignore=shutil.ignore_patterns(
                "__pycache__",
                ".ipynb_checkpoints",
                "*.pyc"
            )
        )
        print("COPIED FOLDER:", folder_name)

for filename in files_to_copy:
    source = PROJECT_ROOT / filename

    if source.exists() and source.is_file():
        shutil.copy2(source, submission_dir / filename)
        print("COPIED FILE:", filename)

# Copy processed dataset
processed_dir = PROJECT_ROOT / "data" / "processed"

if processed_dir.exists():
    shutil.copytree(
        processed_dir,
        submission_dir / "data" / "processed",
        dirs_exist_ok=True
    )
    print("COPIED FOLDER: data/processed")

# Copy selected outputs
output_folders = [
    "outputs/forecasts",
    "outputs/tables",
    "outputs/figures",
    "outputs/uncertainty_analysis",
    "outputs/deployment",
    "outputs/final_quality_assurance"
]

for relative_folder in output_folders:
    source = PROJECT_ROOT / relative_folder

    if source.exists() and source.is_dir():
        destination = submission_dir / relative_folder
        shutil.copytree(
            source,
            destination,
            dirs_exist_ok=True,
            ignore=shutil.ignore_patterns("__pycache__", "*.pyc")
        )
        print("COPIED OUTPUT:", relative_folder)

# Include final presentation
shutil.copy2(
    presentation_path,
    submission_dir / presentation_path.name
)

# Create submission manifest
manifest = []

for path in submission_dir.rglob("*"):
    if path.is_file():
        file_hash = hashlib.sha256(path.read_bytes()).hexdigest()

        manifest.append({
            "file": str(path.relative_to(submission_dir)),
            "size_bytes": path.stat().st_size,
            "sha256": file_hash
        })

manifest_path = submission_dir / "submission_manifest.csv"

pd.DataFrame(manifest).to_csv(
    manifest_path,
    index=False
)

# Create archive
archive_dir = PROJECT_ROOT / "outputs" / "final_archives"
archive_dir.mkdir(parents=True, exist_ok=True)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

archive_path = (
    archive_dir / f"final_submission_{timestamp}.zip"
)

with zipfile.ZipFile(
    archive_path,
    "w",
    compression=zipfile.ZIP_DEFLATED
) as archive:

    for path in submission_dir.rglob("*"):
        if path.is_file():
            archive.write(
                path,
                Path("final_submission") / path.relative_to(submission_dir)
            )

# Verify archive
with zipfile.ZipFile(archive_path, "r") as archive:
    bad_file = archive.testzip()
    archive_count = len(archive.namelist())

print("\n" + "=" * 75)
print("FINAL SUBMISSION SUMMARY")
print("=" * 75)

print("Presentation:", presentation_path.exists())
print("Submission folder:", submission_dir.exists())
print("Submission files:", len(manifest))
print("Manifest:", manifest_path.exists())
print("Archive:", archive_path)
print("Archive entries:", archive_count)
print("Archive integrity:", "PASS" if bad_file is None else "FAIL")

print("\nFINAL SUBMISSION PREPARATION COMPLETED")
````

````python

# ============================================================
# FINAL PROJECT VERIFICATION
# ============================================================
# This section performs a consolidated quality check of the
# forecasting project before final submission.
#
# It verifies documentation, notebook syntax, processed data,
# trained model artifacts, output files, dashboard source code,
# research documents, presentation, and submission archive.
#
# All verification results are saved in the project's
# final_quality_assurance directory.
# ============================================================

from pathlib import Path
import os
import json
import ast
import zipfile
import warnings
import pandas as pd
from datetime import datetime

warnings.filterwarnings("ignore")

PROJECT_ROOT = Path(
    r"C:\Users\hp\Predictive Forecasting of Care Load & Placement Demand"
)

if not PROJECT_ROOT.exists():
    raise FileNotFoundError(f"Project folder not found: {PROJECT_ROOT}")

os.chdir(PROJECT_ROOT)

QA_DIR = PROJECT_ROOT / "outputs" / "final_quality_assurance"
QA_DIR.mkdir(parents=True, exist_ok=True)

checks = []

def record(category, item, status, details=""):
    """Store the result of an individual project verification."""
    checks.append({
        "Category": category,
        "Item": str(item),
        "Status": status,
        "Details": str(details)
    })

def verify_file(category, relative_path):
    """Verify that a required file exists and is not empty."""
    path = PROJECT_ROOT / relative_path

    if path.exists() and path.is_file() and path.stat().st_size > 0:
        record(
            category,
            relative_path,
            "PASS",
            f"{path.stat().st_size:,} bytes"
        )
    else:
        record(category, relative_path, "FAIL", "Missing or empty file")


print("=" * 75)
print("FINAL PROJECT VERIFICATION")
print("=" * 75)


# ============================================================
# DOCUMENTATION VERIFICATION
# ============================================================
# Confirms that the main project documentation files exist
# and contain saved content required for project handover.
# ============================================================

print("\nDOCUMENTATION VERIFICATION")

documentation = [
    "README.md",
    "INSTALLATION_GUIDE.md",
    "USER_GUIDE.md",
    "TECHNICAL_DOCUMENTATION.md"
]

for file in documentation:
    verify_file("Documentation", file)


# ============================================================
# JUPYTER NOTEBOOK VERIFICATION
# ============================================================
# Checks that project notebooks can be opened and that every
# Python code cell has valid Python syntax.
#
# This is a syntax check, not a complete notebook execution test.
# ============================================================

print("\nNOTEBOOK VERIFICATION")

notebook_files = sorted(
    (PROJECT_ROOT / "notebooks").glob("*.ipynb")
)

try:
    import nbformat

    for notebook_path in notebook_files:
        try:
            notebook = nbformat.read(
                notebook_path,
                as_version=4
            )

            for cell in notebook.cells:
                if cell.cell_type == "code":
                    ast.parse(cell.source)

            record(
                "Notebooks",
                notebook_path.relative_to(PROJECT_ROOT),
                "PASS",
                f"{len(notebook.cells)} cells; code syntax valid"
            )

        except Exception as error:
            record(
                "Notebooks",
                notebook_path.relative_to(PROJECT_ROOT),
                "FAIL",
                error
            )

except ImportError as error:
    record("Notebooks", "Notebook verification dependency", "FAIL", error)

if not notebook_files:
    record(
        "Notebooks",
        "Notebook directory",
        "FAIL",
        "No notebooks found"
    )


# ============================================================
# PROCESSED DATASET VERIFICATION
# ============================================================
# Validates the cleaned forecasting dataset.
#
# Checks required columns, date validity, duplicate dates,
# chronological ordering, and completeness of the main target.
# ============================================================

print("\nDATASET VERIFICATION")

dataset_path = (
    PROJECT_ROOT /
    "data" /
    "processed" /
    "cleaned_care_data.csv"
)

required_columns = [
    "date",
    "children_apprehended_and_placed_in_cbp_custody",
    "children_in_cbp_custody",
    "children_transferred_out_of_cbp_custody",
    "children_in_hhs_care",
    "children_discharged_from_hhs_care"
]

try:
    df = pd.read_csv(dataset_path)

    missing_columns = [
        column for column in required_columns
        if column not in df.columns
    ]

    record(
        "Dataset",
        "Required columns",
        "PASS" if not missing_columns else "FAIL",
        "All six columns found" if not missing_columns
        else f"Missing columns: {missing_columns}"
    )

    if "date" in df.columns:
        df["date"] = pd.to_datetime(
            df["date"],
            errors="coerce"
        )

        invalid_dates = int(df["date"].isna().sum())

        record(
            "Dataset",
            "Date validity",
            "PASS" if invalid_dates == 0 else "FAIL",
            f"Invalid dates: {invalid_dates}"
        )

        duplicate_dates = int(df["date"].duplicated().sum())

        record(
            "Dataset",
            "Duplicate dates",
            "PASS" if duplicate_dates == 0 else "FAIL",
            f"Duplicate dates: {duplicate_dates}"
        )

        chronological = df["date"].is_monotonic_increasing

        record(
            "Dataset",
            "Chronological order",
            "PASS" if chronological else "FAIL",
            f"Rows: {len(df)}; date range: "
            f"{df['date'].min()} to {df['date'].max()}"
        )

    if "children_in_hhs_care" in df.columns:
        missing_target = int(
            df["children_in_hhs_care"].isna().sum()
        )

        record(
            "Dataset",
            "Target completeness",
            "PASS" if missing_target == 0 else "FAIL",
            f"Missing target values: {missing_target}"
        )

except Exception as error:
    record(
        "Dataset",
        "Processed dataset",
        "FAIL",
        error
    )


# ============================================================
# TRAINED MODEL ARTIFACT VERIFICATION
# ============================================================
# Locates saved Joblib and Pickle model artifacts.
#
# Each discovered artifact is loaded to check whether it can
# be read by the current Python environment.
# ============================================================

print("\nMODEL VERIFICATION")

model_files = []

for folder in ["models", "outputs"]:
    base = PROJECT_ROOT / folder

    if base.exists():
        model_files.extend([
            path for path in base.rglob("*")
            if path.is_file()
            and path.suffix.lower() in [".joblib", ".pkl"]
            and ".venv" not in str(path)
            and ".deployment_test_env" not in str(path)
        ])

if model_files:
    try:
        import joblib

        for model_path in model_files:
            try:
                joblib.load(model_path)

                record(
                    "Models",
                    model_path.relative_to(PROJECT_ROOT),
                    "PASS",
                    "Artifact loaded successfully"
                )

            except Exception as error:
                record(
                    "Models",
                    model_path.relative_to(PROJECT_ROOT),
                    "FAIL",
                    str(error)[:300]
                )

    except ImportError as error:
        record(
            "Models",
            "Joblib dependency",
            "FAIL",
            error
        )
else:
    record(
        "Models",
        "Model artifacts",
        "FAIL",
        "No model artifacts found"
    )


# ============================================================
# FORECAST AND ANALYSIS OUTPUT VERIFICATION
# ============================================================
# Checks that expected output directories contain files.
#
# CSV and JSON files are opened to verify their readability.
# Empty files and unreadable outputs are recorded as failures.
# ============================================================

print("\nOUTPUT FILE VERIFICATION")

output_folders = [
    "outputs/forecasts",
    "outputs/tables",
    "outputs/figures",
    "outputs/uncertainty_analysis",
    "outputs/deployment"
]

for folder in output_folders:
    base = PROJECT_ROOT / folder

    if not base.exists():
        record(
            "Outputs",
            folder,
            "FAIL",
            "Folder missing"
        )
        continue

    files = [
        path for path in base.rglob("*")
        if path.is_file()
    ]

    record(
        "Outputs",
        folder,
        "PASS" if files else "FAIL",
        f"{len(files)} files found"
    )

    for file_path in files:
        try:
            if file_path.stat().st_size == 0:
                record(
                    "Output files",
                    file_path.relative_to(PROJECT_ROOT),
                    "FAIL",
                    "Empty file"
                )
                continue

            if file_path.suffix.lower() == ".csv":
                pd.read_csv(file_path)

            elif file_path.suffix.lower() == ".json":
                with open(
                    file_path,
                    "r",
                    encoding="utf-8"
                ) as file:
                    json.load(file)

            record(
                "Output files",
                file_path.relative_to(PROJECT_ROOT),
                "PASS",
                "File readable"
            )

        except Exception as error:
            record(
                "Output files",
                file_path.relative_to(PROJECT_ROOT),
                "FAIL",
                str(error)[:300]
            )


# ============================================================
# STREAMLIT DASHBOARD VERIFICATION
# ============================================================
# Checks the dashboard application's Python syntax and searches
# for source-code evidence of the four main dashboard modules.
#
# This source check does not replace a live browser interaction test.
# ============================================================

print("\nDASHBOARD VERIFICATION")

app_path = PROJECT_ROOT / "app.py"

try:
    source = app_path.read_text(encoding="utf-8")

    ast.parse(source)

    record(
        "Dashboard",
        "app.py syntax",
        "PASS",
        "Python syntax valid"
    )

    modules = {
        "Historical monitoring": ["historical", "history"],
        "Predictive forecasting": ["forecast"],
        "Capacity and surge risk": ["capacity", "surge"],
        "Scenario and model comparison": ["scenario", "model"]
    }

    for module, keywords in modules.items():
        found = any(
            word in source.lower()
            for word in keywords
        )

        record(
            "Dashboard",
            module,
            "PASS" if found else "FAIL",
            "Source keyword evidence found"
            if found else "Keyword evidence missing"
        )

except Exception as error:
    record(
        "Dashboard",
        "app.py",
        "FAIL",
        error
    )


# ============================================================
# RESEARCH PAPER AND EXECUTIVE SUMMARY VERIFICATION
# ============================================================
# Confirms that the final research paper and government
# executive summary are available in both DOCX and PDF formats.
# ============================================================

print("\nREPORT VERIFICATION")

report_files = [
    "research_paper/Predictive_Forecasting_Research_Paper.docx",
    "research_paper/Predictive_Forecasting_Research_Paper.pdf",
    "government_executive_summary/Government_Executive_Summary.docx",
    "government_executive_summary/Government_Executive_Summary.pdf"
]

for file in report_files:
    verify_file("Reports", file)


# ============================================================
# FINAL PRESENTATION VERIFICATION
# ============================================================
# Opens the PowerPoint presentation and confirms that it
# contains readable slide structures.
# ============================================================

print("\nPRESENTATION VERIFICATION")

presentation_path = (
    PROJECT_ROOT /
    "final_presentation" /
    "Predictive_Forecasting_Final_Presentation.pptx"
)

try:
    from pptx import Presentation

    presentation = Presentation(str(presentation_path))

    record(
        "Presentation",
        presentation_path.relative_to(PROJECT_ROOT),
        "PASS",
        f"{len(presentation.slides)} slides"
    )

except Exception as error:
    record(
        "Presentation",
        "PowerPoint presentation",
        "FAIL",
        error
    )


# ============================================================
# FINAL SUBMISSION ARCHIVE VERIFICATION
# ============================================================
# Locates the latest final submission ZIP archive.
#
# Tests the ZIP structure and checks whether any archived
# file is corrupted.
# ============================================================

print("\nSUBMISSION ARCHIVE VERIFICATION")

archive_folder = (
    PROJECT_ROOT /
    "outputs" /
    "final_archives"
)

archives = sorted(
    archive_folder.glob("final_submission_*.zip"),
    key=lambda path: path.stat().st_mtime,
    reverse=True
) if archive_folder.exists() else []

if archives:
    archive_path = archives[0]

    try:
        with zipfile.ZipFile(
            archive_path,
            "r"
        ) as archive:

            bad_file = archive.testzip()
            entry_count = len(archive.namelist())

        record(
            "Submission",
            archive_path.name,
            "PASS" if bad_file is None else "FAIL",
            f"ZIP entries: {entry_count}; "
            f"corrupt entry: {bad_file}"
        )

    except Exception as error:
        record(
            "Submission",
            archive_path.name,
            "FAIL",
            error
        )
else:
    record(
        "Submission",
        "Final ZIP archive",
        "FAIL",
        "Archive not found"
    )


# ============================================================
# CONSOLIDATED QUALITY ASSURANCE REPORT
# ============================================================
# Combines all individual verification results into one
# checklist and a machine-readable JSON summary.
#
# The overall status is PASS only when every recorded check
# has passed.
# ============================================================

results = pd.DataFrame(checks)

passed = int(
    (results["Status"] == "PASS").sum()
)

failed = int(
    (results["Status"] == "FAIL").sum()
)

total = len(results)

checklist_path = (
    QA_DIR /
    "final_project_verification_checklist.csv"
)

summary_path = (
    QA_DIR /
    "final_project_verification_summary.json"
)

results.to_csv(
    checklist_path,
    index=False
)

summary = {
    "verification_date": datetime.now().isoformat(
        timespec="seconds"
    ),
    "total_checks": total,
    "passed": passed,
    "failed": failed,
    "overall_status": (
        "PASS" if failed == 0 else "REVIEW REQUIRED"
    ),
    "failure_details": results.loc[
        results["Status"] == "FAIL"
    ].to_dict(orient="records")
}

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        summary,
        file,
        indent=4,
        default=str
    )


# ============================================================
# FINAL VERIFICATION RESULTS
# ============================================================
# Displays category-wise results, total PASS/FAIL counts,
# and the location of the saved verification reports.
# ============================================================

print("\n" + "=" * 75)
print("FINAL QUALITY ASSURANCE CHECKLIST")
print("=" * 75)

print(
    results.groupby(
        ["Category", "Status"]
    ).size().to_string()
)

print("\n" + "=" * 75)
print("FINAL PROJECT STATUS")
print("=" * 75)

print(f"Total checks : {total}")
print(f"PASS         : {passed}")
print(f"FAIL         : {failed}")

print(
    f"Status       : {summary['overall_status']}"
)

if failed:
    print("\nITEMS REQUIRING ATTENTION")

    print(
        results.loc[
            results["Status"] == "FAIL",
            ["Category", "Item", "Details"]
        ].to_string(index=False)
    )
else:
    print("\nALL EXECUTED VERIFICATION CHECKS PASSED")

print("\nSAVED VERIFICATION REPORTS")
print(checklist_path)
print(summary_path)
````
