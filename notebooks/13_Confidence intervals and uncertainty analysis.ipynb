```python

# Confidence Interval Generation and Uncertainty Analysis

import os
import re
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, Markdown

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# PROJECT CONFIGURATION
# ------------------------------------------------------------

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

TARGET_COLUMN = "children_in_hhs_care"
DATE_COLUMN = "date"

FORECAST_HORIZONS = [1, 7, 14, 30]

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

OUTPUT_DIR = PROJECT_ROOT / "outputs"
FORECAST_DIR = OUTPUT_DIR / "forecasts"
EVALUATION_DIR = OUTPUT_DIR / "evaluation"
UNCERTAINTY_DIR = OUTPUT_DIR / "uncertainty_analysis"

for directory in [
    FORECAST_DIR,
    EVALUATION_DIR,
    UNCERTAINTY_DIR
]:
    directory.mkdir(parents=True, exist_ok=True)

print("=" * 75)
print("CONFIDENCE INTERVALS AND UNCERTAINTY ANALYSIS")
print("=" * 75)

print("Project directory:", PROJECT_ROOT)
print("Uncertainty output directory:", UNCERTAINTY_DIR)


# ------------------------------------------------------------
# LOAD EXISTING PREDICTIONS
# ------------------------------------------------------------

def normalize_column_names(df):

    df = df.copy()

    df.columns = [
        str(col).strip().replace(" ", "_")
        for col in df.columns
    ]

    return df


def find_column(df, candidates):

    lookup = {
        str(col).lower(): col
        for col in df.columns
    }

    for candidate in candidates:

        if candidate.lower() in lookup:
            return lookup[candidate.lower()]

    return None


def prepare_predictions(df, default_partition=None):

    df = normalize_column_names(df)

    aliases = {
        "Model": [
            "Model", "model", "model_name",
            "algorithm", "estimator"
        ],

        "Actual": [
            "Actual", "actual", "y_true",
            "actual_value", "actuals"
        ],

        "Predicted": [
            "Predicted", "predicted", "y_pred",
            "prediction", "forecast", "predicted_value"
        ],

        "Horizon": [
            "Horizon", "horizon", "forecast_horizon",
            "horizon_days", "step"
        ],

        "Partition": [
            "Partition", "partition", "dataset",
            "split", "data_partition"
        ],

        "Actual_Target_Date": [
            "Actual_Target_Date", "actual_target_date",
            "target_date", "date", "Date"
        ],

        "Period_Type": [
            "Period_Type", "period_type",
            "period", "regime"
        ],

        "Forecast_Origin": [
            "Forecast_Origin", "forecast_origin",
            "origin_date"
        ]
    }

    rename = {}

    for standard, possibilities in aliases.items():

        found = find_column(df, possibilities)

        if found is not None and found != standard:
            rename[found] = standard

    df = df.rename(columns=rename)

    required = ["Actual", "Predicted"]

    missing = [
        col for col in required
        if col not in df.columns
    ]

    if missing:
        raise ValueError(
            f"Prediction file is missing required columns: {missing}. "
            f"Available columns: {df.columns.tolist()}"
        )

    if "Model" not in df.columns:
        df["Model"] = "Unknown_Model"

    if "Horizon" not in df.columns:
        df["Horizon"] = 1

    if "Partition" not in df.columns:

        if default_partition is None:
            df["Partition"] = "unknown"
        else:
            df["Partition"] = default_partition

    if "Actual_Target_Date" not in df.columns:
        df["Actual_Target_Date"] = pd.NaT

    if "Period_Type" not in df.columns:
        df["Period_Type"] = "Unknown"

    if "Forecast_Origin" not in df.columns:
        df["Forecast_Origin"] = pd.NaT

    df["Actual"] = pd.to_numeric(
        df["Actual"], errors="coerce"
    )

    df["Predicted"] = pd.to_numeric(
        df["Predicted"], errors="coerce"
    )

    df["Horizon"] = pd.to_numeric(
        df["Horizon"], errors="coerce"
    ).fillna(1).astype(int)

    df["Actual_Target_Date"] = pd.to_datetime(
        df["Actual_Target_Date"], errors="coerce"
    )

    df["Forecast_Origin"] = pd.to_datetime(
        df["Forecast_Origin"], errors="coerce"
    )

    df["Model"] = (
        df["Model"].astype(str).str.strip()
    )

    df["Partition"] = (
        df["Partition"].astype(str).str.lower().str.strip()
    )

    df = df.dropna(
        subset=["Actual", "Predicted"]
    ).copy()

    df["Residual"] = (
        df["Actual"] - df["Predicted"]
    )

    df["Absolute_Error"] = df["Residual"].abs()

    return df.reset_index(drop=True)


# Collect predictions from existing notebook variables.

prediction_sources = []

for variable_name, partition_name in [
    ("validation_predictions", "validation"),
    ("test_predictions", "testing"),
    ("predictions", None)
]:

    if variable_name in globals():

        obj = globals()[variable_name]

        if isinstance(obj, pd.DataFrame) and not obj.empty:

            prediction_sources.append(
                prepare_predictions(
                    obj,
                    default_partition=partition_name
                )
            )

            print(
                f"Loaded notebook variable: {variable_name}",
                obj.shape
            )


# Search saved prediction files.

search_directories = [
    OUTPUT_DIR,
    FORECAST_DIR,
    EVALUATION_DIR,
    PROJECT_ROOT / "data" / "processed",
    PROJECT_ROOT / "data" / "predictions"
]

candidate_files = []

for directory in search_directories:

    if directory.exists():

        candidate_files.extend(
            directory.rglob("*.csv")
        )

candidate_files = list(dict.fromkeys(candidate_files))

loaded_files = []

for file_path in candidate_files:

    filename = file_path.name.lower()

    if any(
        word in filename
        for word in [
            "prediction",
            "forecast",
            "validation",
            "testing",
            "test"
        ]
    ):

        try:

            temp = pd.read_csv(file_path)

            if (
                find_column(
                    temp,
                    ["Actual", "actual", "y_true"]
                ) is not None
                and find_column(
                    temp,
                    ["Predicted", "predicted", "y_pred",
                     "forecast", "prediction"]
                ) is not None
            ):

                if "validation" in filename:
                    default = "validation"

                elif "test" in filename:
                    default = "testing"

                else:
                    default = None

                loaded_files.append(
                    prepare_predictions(
                        temp,
                        default_partition=default
                    )
                )

                print("Loaded prediction file:", file_path)

        except Exception:
            continue


# Combine available prediction sources.

all_sources = prediction_sources + loaded_files

if not all_sources:

    raise FileNotFoundError(
        "No usable prediction data was found. "
        "Run the forecasting and model evaluation notebooks first. "
        "Expected columns: Model, Actual, Predicted, Horizon, Partition."
    )

predictions_df = pd.concat(
    all_sources,
    ignore_index=True
)

# Remove duplicate prediction records.

dedup_columns = [
    "Model",
    "Actual",
    "Predicted",
    "Horizon",
    "Partition",
    "Actual_Target_Date"
]

predictions_df = predictions_df.drop_duplicates(
    subset=dedup_columns
).reset_index(drop=True)

print("\nCombined predictions:", predictions_df.shape)

print("\nAvailable partitions:")
print(predictions_df["Partition"].value_counts())

print("\nAvailable models:")
print(predictions_df["Model"].unique())


# ------------------------------------------------------------
# IDENTIFY VALIDATION AND TESTING DATA
# ------------------------------------------------------------

validation_labels = [
    "validation",
    "valid",
    "val"
]

testing_labels = [
    "testing",
    "test",
    "testing_set",
    "test_set"
]

validation_df = predictions_df[
    predictions_df["Partition"].isin(validation_labels)
].copy()

test_df = predictions_df[
    predictions_df["Partition"].isin(testing_labels)
].copy()

if validation_df.empty:

    raise ValueError(
        "Validation predictions are unavailable. "
        "Residual bootstrap requires independent validation residuals."
    )

if test_df.empty:

    raise ValueError(
        "Testing predictions are unavailable. "
        "Generate testing predictions before evaluating intervals."
    )

print("\nValidation observations:", len(validation_df))
print("Testing observations:", len(test_df))


# ------------------------------------------------------------
# CALCULATE VALIDATION RESIDUAL DISTRIBUTIONS
# ------------------------------------------------------------

validation_df["Residual"] = (
    validation_df["Actual"] -
    validation_df["Predicted"]
)

global_residuals = (
    validation_df["Residual"]
    .dropna()
    .to_numpy()
)

if len(global_residuals) < 5:

    raise ValueError(
        "Insufficient validation residuals for reliable "
        "bootstrap interval estimation."
    )


def get_residual_distribution(model, horizon):

    # First preference: model-specific and horizon-specific.

    selected = validation_df[
        (validation_df["Model"] == model) &
        (validation_df["Horizon"] == horizon)
    ]["Residual"].dropna().to_numpy()

    if len(selected) >= 5:
        return selected, "Model_Horizon_Residual_Bootstrap"

    # Second preference: model-specific.

    selected = validation_df[
        validation_df["Model"] == model
    ]["Residual"].dropna().to_numpy()

    if len(selected) >= 5:
        return selected, "Model_Residual_Bootstrap"

    # Third preference: horizon-specific.

    selected = validation_df[
        validation_df["Horizon"] == horizon
    ]["Residual"].dropna().to_numpy()

    if len(selected) >= 5:
        return selected, "Horizon_Residual_Bootstrap"

    # Final fallback: global validation residuals.

    return (
        global_residuals,
        "Global_Validation_Residual_Bootstrap"
    )


# ------------------------------------------------------------
# GENERATE LOWER AND UPPER PREDICTION BOUNDS
# ------------------------------------------------------------

def generate_prediction_intervals(df):

    result = df.copy()

    result["Lower_80"] = np.nan
    result["Upper_80"] = np.nan

    result["Lower_95"] = np.nan
    result["Upper_95"] = np.nan

    result["Interval_Method"] = None

    result["Residual_Sample_Size"] = 0

    result["Lower_80"] = result["Lower_80"].astype(float)
    result["Upper_80"] = result["Upper_80"].astype(float)
    result["Lower_95"] = result["Lower_95"].astype(float)
    result["Upper_95"] = result["Upper_95"].astype(float)

    result["Residual_Sample_Size"] = (
        result["Residual_Sample_Size"].astype(int)
    )

    for index, row in result.iterrows():

        residuals, method = get_residual_distribution(
            row["Model"],
            row["Horizon"]
        )

        # Residual bootstrap.
        # Empirical residual quantiles produce asymmetric bounds.

        q025 = np.quantile(residuals, 0.025)
        q10 = np.quantile(residuals, 0.10)
        q90 = np.quantile(residuals, 0.90)
        q975 = np.quantile(residuals, 0.975)

        prediction = float(row["Predicted"])

        result.at[index, "Lower_80"] = max(
            0, prediction + q10
        )

        result.at[index, "Upper_80"] = max(
            0, prediction + q90
        )

        result.at[index, "Lower_95"] = max(
            0, prediction + q025
        )

        result.at[index, "Upper_95"] = max(
            0, prediction + q975
        )

        result.at[index, "Interval_Method"] = method

        result.at[index, "Residual_Sample_Size"] = len(
            residuals
        )

    # Calculate interval widths.

    result["Width_80"] = (
        result["Upper_80"] - result["Lower_80"]
    )

    result["Width_95"] = (
        result["Upper_95"] - result["Lower_95"]
    )

    return result


# Generate intervals for testing and validation predictions.

test_intervals = generate_prediction_intervals(
    test_df.reset_index(drop=True)
)

validation_intervals = generate_prediction_intervals(
    validation_df.reset_index(drop=True)
)

# Testing data is the primary independent evaluation dataset.

interval_predictions = pd.concat(
    [validation_intervals, test_intervals],
    ignore_index=True
)

interval_predictions = interval_predictions.sort_values(
    ["Partition", "Model", "Horizon"]
).reset_index(drop=True)

print("\nPrediction intervals generated successfully.")

display(
    interval_predictions[
        [
            "Model",
            "Actual",
            "Predicted",
            "Horizon",
            "Partition",
            "Lower_80",
            "Upper_80",
            "Lower_95",
            "Upper_95",
            "Interval_Method"
        ]
    ].head(10)
)


# ------------------------------------------------------------
# EMPIRICAL INTERVAL COVERAGE
# ------------------------------------------------------------

def calculate_coverage(df):

    result = df.copy()

    result["Covered_80"] = (
        (result["Actual"] >= result["Lower_80"]) &
        (result["Actual"] <= result["Upper_80"])
    )

    result["Covered_95"] = (
        (result["Actual"] >= result["Lower_95"]) &
        (result["Actual"] <= result["Upper_95"])
    )

    result["Coverage_Error_80"] = (
        result["Covered_80"].astype(int) - 0.80
    )

    result["Coverage_Error_95"] = (
        result["Covered_95"].astype(int) - 0.95
    )

    return result


interval_predictions = calculate_coverage(
    interval_predictions
)

test_intervals = calculate_coverage(
    test_intervals
)


def summarize_intervals(df, group_columns):

    records = []

    grouped = df.groupby(
        group_columns,
        dropna=False
    )

    for group_key, group in grouped:

        if not isinstance(group_key, tuple):
            group_key = (group_key,)

        record = dict(
            zip(group_columns, group_key)
        )

        record.update({

            "Sample_Size": len(group),

            "Coverage_80": (
                group["Covered_80"].mean()
            ),

            "Coverage_95": (
                group["Covered_95"].mean()
            ),

            "Mean_Width_80": (
                group["Width_80"].mean()
            ),

            "Mean_Width_95": (
                group["Width_95"].mean()
            ),

            "Median_Width_80": (
                group["Width_80"].median()
            ),

            "Median_Width_95": (
                group["Width_95"].median()
            ),

            "Target_Coverage_80": 0.80,

            "Target_Coverage_95": 0.95,

            "Coverage_Error_80": abs(
                group["Covered_80"].mean() - 0.80
            ),

            "Coverage_Error_95": abs(
                group["Covered_95"].mean() - 0.95
            )

        })

        records.append(record)

    return pd.DataFrame(records)


coverage_metrics = summarize_intervals(
    test_intervals,
    ["Model", "Horizon"]
)

print("\nTesting interval coverage by model and horizon:")

display(coverage_metrics)


# ------------------------------------------------------------
# ANALYZE UNCERTAINTY BY FORECAST HORIZON
# ------------------------------------------------------------

horizon_uncertainty = summarize_intervals(
    test_intervals,
    ["Horizon"]
)

horizon_uncertainty = horizon_uncertainty.sort_values(
    "Horizon"
)

print("\nUncertainty by forecast horizon:")

display(horizon_uncertainty)


# ------------------------------------------------------------
# IDENTIFY SURGE AND STABLE PERIODS
# ------------------------------------------------------------

actual_periods = interval_predictions.copy()

if actual_periods["Actual_Target_Date"].notna().any():

    actual_periods = actual_periods.sort_values(
        "Actual_Target_Date"
    )

    actual_series = (
        actual_periods
        .drop_duplicates("Actual_Target_Date")
        .set_index("Actual_Target_Date")["Actual"]
        .sort_index()
    )

    # Change between observed reporting records.
    # No missing calendar dates are artificially filled.

    actual_change = actual_series.pct_change().abs()

    actual_change = actual_change.replace(
        [np.inf, -np.inf], np.nan
    )

    actual_periods["Absolute_Actual_Change"] = (
        actual_periods["Actual_Target_Date"]
        .map(actual_change)
    )

else:

    actual_periods["Absolute_Actual_Change"] = np.nan

# Respect an existing period classification when available.

existing_period = (
    actual_periods["Period_Type"]
    .astype(str)
    .str.lower()
)

if existing_period.isin(
    ["surge", "stable"]
).any():

    actual_periods["Period_Category"] = np.where(
        existing_period.eq("surge"),
        "Surge",
        np.where(
            existing_period.eq("stable"),
            "Stable",
            "Unclassified"
        )
    )

else:

    valid_changes = (
        actual_periods["Absolute_Actual_Change"]
        .dropna()
    )

    if len(valid_changes) > 0:

        surge_threshold = valid_changes.quantile(0.75)

        stable_threshold = valid_changes.quantile(0.50)

        actual_periods["Period_Category"] = np.select(
            [
                actual_periods[
                    "Absolute_Actual_Change"
                ] >= surge_threshold,

                actual_periods[
                    "Absolute_Actual_Change"
                ] <= stable_threshold
            ],
            [
                "Surge",
                "Stable"
            ],
            default="Normal"
        )

    else:

        actual_periods["Period_Category"] = "Unclassified"


# ------------------------------------------------------------
# UNCERTAINTY DURING SURGES AND STABLE PERIODS
# ------------------------------------------------------------

period_uncertainty = summarize_intervals(
    actual_periods[
        actual_periods["Period_Category"].isin(
            ["Surge", "Stable"]
        )
    ],
    ["Period_Category"]
)

print("\nUncertainty during surge and stable periods:")

display(period_uncertainty)


# ------------------------------------------------------------
# UNCERTAINTY BY MODEL, HORIZON AND PERIOD TYPE
# ------------------------------------------------------------

period_horizon_uncertainty = summarize_intervals(
    actual_periods[
        actual_periods["Period_Category"].isin(
            ["Surge", "Stable"]
        )
    ],
    ["Model", "Horizon", "Period_Category"]
)

print("\nDetailed uncertainty by model, horizon and period:")

display(period_horizon_uncertainty)


# ------------------------------------------------------------
# INTERVAL CALIBRATION ASSESSMENT
# ------------------------------------------------------------

calibration_metrics = coverage_metrics.copy()

calibration_metrics["Calibration_Status_80"] = np.select(
    [
        calibration_metrics["Coverage_80"].between(
            0.75, 0.85
        ),

        calibration_metrics["Coverage_80"] < 0.75
    ],
    [
        "Well Calibrated",
        "Undercoverage"
    ],
    default="Overcoverage"
)

calibration_metrics["Calibration_Status_95"] = np.select(
    [
        calibration_metrics["Coverage_95"].between(
            0.90, 1.00
        ),

        calibration_metrics["Coverage_95"] < 0.90
    ],
    [
        "Well Calibrated",
        "Undercoverage"
    ],
    default="Overcoverage"
)

print("\nInterval calibration assessment:")

display(calibration_metrics)


# ------------------------------------------------------------
# BOOTSTRAP UNCERTAINTY DIAGNOSTICS
# ------------------------------------------------------------

bootstrap_diagnostics = (
    test_intervals.groupby(
        ["Model", "Horizon", "Interval_Method"],
        dropna=False
    )
    .agg(
        Prediction_Count=("Predicted", "count"),

        Mean_Residual_Sample_Size=(
            "Residual_Sample_Size", "mean"
        ),

        Mean_Absolute_Error=(
            "Absolute_Error", "mean"
        ),

        Mean_80_Interval_Width=(
            "Width_80", "mean"
        ),

        Mean_95_Interval_Width=(
            "Width_95", "mean"
        )
    )
    .reset_index()
)

print("\nResidual bootstrap diagnostics:")

display(bootstrap_diagnostics)


# ------------------------------------------------------------
# SAVE ALL UNCERTAINTY RESULTS
# ------------------------------------------------------------

interval_predictions.to_csv(
    UNCERTAINTY_DIR / "confidence_intervals.csv",
    index=False
)

test_intervals.to_csv(
    UNCERTAINTY_DIR / "testing_confidence_intervals.csv",
    index=False
)

validation_intervals.to_csv(
    UNCERTAINTY_DIR / "validation_confidence_intervals.csv",
    index=False
)

coverage_metrics.to_csv(
    UNCERTAINTY_DIR / "empirical_interval_coverage.csv",
    index=False
)

horizon_uncertainty.to_csv(
    UNCERTAINTY_DIR / "horizon_uncertainty.csv",
    index=False
)

period_uncertainty.to_csv(
    UNCERTAINTY_DIR / "surge_stable_uncertainty.csv",
    index=False
)

period_horizon_uncertainty.to_csv(
    UNCERTAINTY_DIR / "period_horizon_uncertainty.csv",
    index=False
)

calibration_metrics.to_csv(
    UNCERTAINTY_DIR / "interval_calibration.csv",
    index=False
)

bootstrap_diagnostics.to_csv(
    UNCERTAINTY_DIR / "bootstrap_diagnostics.csv",
    index=False
)

actual_periods.to_csv(
    UNCERTAINTY_DIR / "uncertainty_period_analysis.csv",
    index=False
)


# ------------------------------------------------------------
# VISUALIZATION: PREDICTION INTERVALS
# ------------------------------------------------------------

best_model = (
    test_intervals["Model"]
    .value_counts()
    .index[0]
)

best_horizon = 1

plot_data = test_intervals[
    (test_intervals["Model"] == best_model) &
    (test_intervals["Horizon"] == best_horizon)
].copy()

plot_data = plot_data.reset_index(drop=True)

if not plot_data.empty:

    fig, ax = plt.subplots(figsize=(15, 6))

    x = np.arange(len(plot_data))

    ax.plot(
        x,
        plot_data["Actual"],
        label="Actual",
        linewidth=1.8
    )

    ax.plot(
        x,
        plot_data["Predicted"],
        label="Predicted",
        linewidth=1.5
    )

    ax.fill_between(
        x,
        plot_data["Lower_80"],
        plot_data["Upper_80"],
        alpha=0.25,
        label="80% Prediction Interval"
    )

    ax.fill_between(
        x,
        plot_data["Lower_95"],
        plot_data["Upper_95"],
        alpha=0.12,
        label="95% Prediction Interval"
    )

    ax.set_title(
        f"Prediction Intervals: {best_model} "
        f"({best_horizon}-Day Horizon)"
    )

    ax.set_xlabel("Testing Observation")
    ax.set_ylabel("Children in HHS Care")

    ax.legend()
    ax.grid(alpha=0.25)

    plt.tight_layout()

    plt.savefig(
        UNCERTAINTY_DIR / "prediction_intervals.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# VISUALIZATION: UNCERTAINTY BY HORIZON
# ------------------------------------------------------------

if not horizon_uncertainty.empty:

    fig, ax = plt.subplots(figsize=(10, 5))

    ax.plot(
        horizon_uncertainty["Horizon"],
        horizon_uncertainty["Mean_Width_80"],
        marker="o",
        label="80% Interval Width"
    )

    ax.plot(
        horizon_uncertainty["Horizon"],
        horizon_uncertainty["Mean_Width_95"],
        marker="o",
        label="95% Interval Width"
    )

    ax.set_title("Prediction Uncertainty by Forecast Horizon")
    ax.set_xlabel("Forecast Horizon (Days)")
    ax.set_ylabel("Average Interval Width")

    ax.legend()
    ax.grid(alpha=0.25)

    plt.tight_layout()

    plt.savefig(
        UNCERTAINTY_DIR / "uncertainty_by_horizon.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# VISUALIZATION: EMPIRICAL COVERAGE
# ------------------------------------------------------------

if not coverage_metrics.empty:

    coverage_plot = (
        coverage_metrics.groupby("Horizon")
        [["Coverage_80", "Coverage_95"]]
        .mean()
    )

    fig, ax = plt.subplots(figsize=(10, 5))

    x = np.arange(len(coverage_plot))

    width = 0.35

    ax.bar(
        x - width / 2,
        coverage_plot["Coverage_80"],
        width,
        label="Empirical 80% Coverage"
    )

    ax.bar(
        x + width / 2,
        coverage_plot["Coverage_95"],
        width,
        label="Empirical 95% Coverage"
    )

    ax.axhline(
        0.80,
        linestyle="--",
        label="Target 80%"
    )

    ax.axhline(
        0.95,
        linestyle=":",
        label="Target 95%"
    )

    ax.set_xticks(x)

    ax.set_xticklabels(
        coverage_plot.index.astype(str)
    )

    ax.set_xlabel("Forecast Horizon (Days)")
    ax.set_ylabel("Empirical Coverage")

    ax.set_title("Prediction Interval Coverage")

    ax.legend()
    ax.grid(axis="y", alpha=0.25)

    plt.tight_layout()

    plt.savefig(
        UNCERTAINTY_DIR / "interval_coverage.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# DOCUMENT UNCERTAINTY LIMITATIONS
# ------------------------------------------------------------

limitations = """
CONFIDENCE INTERVALS AND UNCERTAINTY ANALYSIS
==============================================

Methodology:
- Empirical residual bootstrap.
- Validation residuals used for interval calibration.
- Model-specific and horizon-specific residual distributions
  are preferred.
- Model-specific, horizon-specific and global residual
  fallbacks are applied when necessary.
- Asymmetric empirical residual quantiles are used.
- Nonnegative lower prediction bounds are enforced.

Prediction intervals:
- 80% prediction intervals.
- 95% prediction intervals.

Evaluation:
- Empirical interval coverage.
- Prediction interval width.
- Horizon-specific uncertainty.
- Surge-period uncertainty.
- Stable-period uncertainty.
- Interval calibration diagnostics.

Limitations:
1. Prediction intervals are conditional on the forecasting
   model and validation residual distribution.

2. Bootstrap residuals may not be independent. Temporal
   autocorrelation can affect interval reliability.

3. Distribution shifts between validation and testing periods
   can reduce empirical coverage.

4. Extreme surges may be insufficiently represented in the
   validation residual distribution.

5. Small model-horizon residual samples can produce unstable
   interval estimates.

6. The empirical intervals are not guaranteed to achieve
   nominal 80% or 95% coverage.

7. Lower prediction bounds are restricted to zero because
   the number of children in care cannot be negative.

8. Missing reporting dates are not artificially interpolated.

9. Surge and stable classifications are descriptive and
   depend on observed changes or existing period labels.

10. Interval coverage should be interpreted alongside
    interval width, forecasting accuracy and sample size.

11. The bootstrap implementation uses empirical residual
    quantiles; it does not fully model future structural
    changes or unknown external events.

12. Validation-based calibration must be monitored and
    recalibrated when the underlying data distribution changes.
"""

with open(
    UNCERTAINTY_DIR / "uncertainty_limitations.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write(limitations)


# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CONFIDENCE INTERVAL AND UNCERTAINTY ANALYSIS COMPLETED")
print("=" * 75)

print("\nPrediction records:", len(interval_predictions))

print(
    "Validation records:",
    len(validation_intervals)
)

print(
    "Testing records:",
    len(test_intervals)
)

print(
    "Models evaluated:",
    test_intervals["Model"].nunique()
)

print(
    "Forecast horizons:",
    sorted(test_intervals["Horizon"].unique().tolist())
)

print(
    "80% mean empirical coverage:",
    round(test_intervals["Covered_80"].mean() * 100, 2),
    "%"
)

print(
    "95% mean empirical coverage:",
    round(test_intervals["Covered_95"].mean() * 100, 2),
    "%"
)

print(
    "Mean 80% interval width:",
    round(test_intervals["Width_80"].mean(), 2)
)

print(
    "Mean 95% interval width:",
    round(test_intervals["Width_95"].mean(), 2)
)

print("\nSaved files:")

for file_path in sorted(UNCERTAINTY_DIR.iterdir()):

    if file_path.is_file():
        print(" -", file_path.name)

print("\nOutput directory:")
print(UNCERTAINTY_DIR)

print("\nAll uncertainty-analysis calculations and reports generated.")

display(Markdown(
    "**Confidence interval and uncertainty analysis cell executed.**"
))
```
