```python

# Operational KPIs and Capacity Risk Analysis

import os
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, Markdown

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# PROJECT CONFIGURATION
# ------------------------------------------------------------

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

TARGET_COLUMN = "children_in_hhs_care"
DATE_COLUMN = "date"

FORECAST_HORIZONS = [1, 7, 14, 30]

TEST_SIZE = 0.20

# Surge definition: 90th percentile of training-period demand.
SURGE_QUANTILE = 0.90

# Alternative training windows for sensitivity analysis.
TRAINING_WINDOWS = [0.60, 0.80, 0.90]

# Operational capacity must come from a verified source.
# Do not enter an estimated or invented capacity value.
CAPACITY_THRESHOLD = None

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = (
    PROJECT_ROOT /
    "data" /
    "processed" /
    "cleaned_care_data.csv"
)

OUTPUT_DIR = PROJECT_ROOT / "outputs"

UNCERTAINTY_DIR = (
    OUTPUT_DIR / "uncertainty_analysis"
)

KPI_DIR = (
    OUTPUT_DIR / "operational_kpi_analysis"
)

KPI_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("=" * 80)
print("OPERATIONAL KPIs AND CAPACITY RISK ANALYSIS")
print("=" * 80)

print("Project directory:", PROJECT_ROOT)
print("Dataset:", DATA_PATH)
print("Output directory:", KPI_DIR)


# ------------------------------------------------------------
# LOAD AND VALIDATE HISTORICAL DATA
# ------------------------------------------------------------

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Cleaned dataset not found: {DATA_PATH}"
    )

historical_df = pd.read_csv(DATA_PATH)

historical_df.columns = [
    str(col).strip().lower().replace(" ", "_")
    for col in historical_df.columns
]

if DATE_COLUMN not in historical_df.columns:
    raise KeyError(
        f"Date column '{DATE_COLUMN}' not found."
    )

if TARGET_COLUMN not in historical_df.columns:
    raise KeyError(
        f"Target column '{TARGET_COLUMN}' not found."
    )

historical_df[DATE_COLUMN] = pd.to_datetime(
    historical_df[DATE_COLUMN],
    errors="coerce"
)

historical_df[TARGET_COLUMN] = pd.to_numeric(
    historical_df[TARGET_COLUMN],
    errors="coerce"
)

historical_df = (
    historical_df
    .dropna(subset=[DATE_COLUMN])
    .sort_values(DATE_COLUMN)
    .drop_duplicates(subset=[DATE_COLUMN], keep="last")
    .reset_index(drop=True)
)

historical_df = historical_df[
    historical_df[TARGET_COLUMN].notna()
].copy()

if len(historical_df) < 30:
    raise ValueError(
        "Insufficient historical observations for KPI analysis."
    )

print("\nHistorical observations:", len(historical_df))

print(
    "Historical period:",
    historical_df[DATE_COLUMN].min(),
    "to",
    historical_df[DATE_COLUMN].max()
)

print(
    "Missing target observations:",
    historical_df[TARGET_COLUMN].isna().sum()
)


# ------------------------------------------------------------
# LOAD FORECAST PREDICTIONS
# ------------------------------------------------------------

def standardize_predictions(df, default_partition=None):

    df = df.copy()

    df.columns = [
        str(col).strip().replace(" ", "_")
        for col in df.columns
    ]

    aliases = {
        "Model": [
            "Model", "model", "model_name",
            "algorithm", "estimator"
        ],

        "Actual": [
            "Actual", "actual", "y_true",
            "actual_value", "actuals"
        ],

        "Predicted": [
            "Predicted", "predicted", "y_pred",
            "prediction", "forecast", "predicted_value"
        ],

        "Horizon": [
            "Horizon", "horizon",
            "forecast_horizon", "horizon_days", "step"
        ],

        "Partition": [
            "Partition", "partition",
            "dataset", "split", "data_partition"
        ],

        "Actual_Target_Date": [
            "Actual_Target_Date", "actual_target_date",
            "target_date", "date", "Date"
        ],

        "Forecast_Origin": [
            "Forecast_Origin", "forecast_origin",
            "origin_date"
        ],

        "Period_Type": [
            "Period_Type", "period_type",
            "period", "regime"
        ],

        "Lower_80": ["Lower_80", "lower_80"],
        "Upper_80": ["Upper_80", "upper_80"],
        "Lower_95": ["Lower_95", "lower_95"],
        "Upper_95": ["Upper_95", "upper_95"]
    }

    rename = {}

    lower_lookup = {
        str(col).lower(): col
        for col in df.columns
    }

    for standard, candidates in aliases.items():

        for candidate in candidates:

            if candidate.lower() in lower_lookup:

                original = lower_lookup[candidate.lower()]

                if original != standard:
                    rename[original] = standard

                break

    df = df.rename(columns=rename)

    if not {"Actual", "Predicted"}.issubset(df.columns):
        return None

    if "Model" not in df.columns:
        df["Model"] = "Unknown_Model"

    if "Horizon" not in df.columns:
        df["Horizon"] = 1

    if "Partition" not in df.columns:
        df["Partition"] = (
            default_partition
            if default_partition is not None
            else "unknown"
        )

    for col in [
        "Actual_Target_Date",
        "Forecast_Origin"
    ]:

        if col not in df.columns:
            df[col] = pd.NaT

        df[col] = pd.to_datetime(
            df[col],
            errors="coerce"
        )

    if "Period_Type" not in df.columns:
        df["Period_Type"] = "Unknown"

    for col in [
        "Actual",
        "Predicted",
        "Horizon",
        "Lower_80",
        "Upper_80",
        "Lower_95",
        "Upper_95"
    ]:

        if col in df.columns:

            df[col] = pd.to_numeric(
                df[col],
                errors="coerce"
            )

    df["Model"] = (
        df["Model"]
        .astype(str)
        .str.strip()
    )

    df["Partition"] = (
        df["Partition"]
        .astype(str)
        .str.lower()
        .str.strip()
    )

    df = df.replace({
        "Model": {
            "nan": np.nan,
            "None": np.nan,
            "": np.nan
        }
    })

    df = df.dropna(
        subset=["Actual", "Predicted", "Model"]
    )

    df["Horizon"] = (
        df["Horizon"]
        .fillna(1)
        .astype(int)
    )

    return df.reset_index(drop=True)


# Prefer the consolidated prediction file.
# Otherwise, load separate validation and testing files.

prediction_candidates = [
    (
        OUTPUT_DIR / "combined_model_predictions.csv",
        None
    ),
    (
        OUTPUT_DIR / "validation_predictions.csv",
        "validation"
    ),
    (
        OUTPUT_DIR / "machine_learning_validation_predictions.csv",
        "validation"
    ),
    (
        OUTPUT_DIR / "statistical_validation_forecasts.csv",
        "validation"
    ),
    (
        OUTPUT_DIR / "evaluation" / "validation_predictions.csv",
        "validation"
    ),
    (
        OUTPUT_DIR / "evaluation" / "test_predictions.csv",
        "testing"
    ),
    (
        OUTPUT_DIR / "machine_learning_test_predictions.csv",
        "testing"
    ),
    (
        OUTPUT_DIR / "statistical_test_forecasts.csv",
        "testing"
    )
]

prediction_frames = []

for file_path, partition in prediction_candidates:

    if not file_path.exists():
        continue

    try:

        temp = pd.read_csv(file_path)

        temp = standardize_predictions(
            temp,
            default_partition=partition
        )

        if temp is not None and not temp.empty:

            prediction_frames.append(temp)

            print(
                "Loaded:",
                file_path.name,
                "| Records:",
                len(temp)
            )

            # Do not mix consolidated and component files.
            if file_path.name == "combined_model_predictions.csv":
                break

    except Exception as error:

        print(
            "Could not load:",
            file_path.name,
            "|",
            error
        )


# Check for prediction data already loaded in the notebook.

if not prediction_frames:

    for variable_name, partition in [
        ("validation_predictions", "validation"),
        ("test_predictions", "testing"),
        ("validation_df", "validation"),
        ("test_df", "testing")
    ]:

        if variable_name in globals():

            obj = globals()[variable_name]

            if isinstance(obj, pd.DataFrame):

                temp = standardize_predictions(
                    obj,
                    default_partition=partition
                )

                if temp is not None and not temp.empty:
                    prediction_frames.append(temp)


if not prediction_frames:

    raise FileNotFoundError(
        "No usable validation/testing prediction files found. "
        "Run the forecasting evaluation notebooks first."
    )

predictions = pd.concat(
    prediction_frames,
    ignore_index=True
)

predictions = predictions.drop_duplicates(
    subset=[
        "Model",
        "Actual",
        "Predicted",
        "Horizon",
        "Partition",
        "Actual_Target_Date",
        "Forecast_Origin"
    ]
).reset_index(drop=True)

validation_df = predictions[
    predictions["Partition"].isin(
        ["validation", "valid", "val"]
    )
].copy()

test_df = predictions[
    predictions["Partition"].isin(
        ["testing", "test", "test_set", "out_of_sample"]
    )
].copy()

if validation_df.empty or test_df.empty:

    raise ValueError(
        "Both validation and testing predictions are required. "
        f"Validation records: {len(validation_df)}; "
        f"Testing records: {len(test_df)}."
    )

print("\nValidation records:", len(validation_df))
print("Testing records:", len(test_df))

print(
    "Available models:",
    sorted(test_df["Model"].unique().tolist())
)


# ------------------------------------------------------------
# DEFINE AND CALCULATE FORECAST ACCURACY KPI
# ------------------------------------------------------------

# Definition:
# Forecast Accuracy KPI = max(0, 1 - WAPE) * 100
#
# WAPE = Sum of absolute forecast errors / Sum of actual values.
#
# This KPI is expressed as a percentage and is not the
# percentage of individual forecasts predicted correctly.

def calculate_wape(actual, predicted):

    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    denominator = np.sum(np.abs(actual))

    if denominator == 0:
        return np.nan

    return (
        np.sum(np.abs(actual - predicted))
        / denominator
    )


def calculate_accuracy_kpi(actual, predicted):

    wape = calculate_wape(
        actual,
        predicted
    )

    if pd.isna(wape):
        return np.nan

    return max(0, 1 - wape) * 100


def calculate_forecast_metrics(df, group_columns):

    records = []

    for group_key, group in df.groupby(
        group_columns,
        dropna=False
    ):

        if not isinstance(group_key, tuple):
            group_key = (group_key,)

        record = dict(
            zip(group_columns, group_key)
        )

        actual = group["Actual"].to_numpy()
        predicted = group["Predicted"].to_numpy()

        errors = actual - predicted

        record.update({

            "Sample_Size": len(group),

            "MAE": np.mean(np.abs(errors)),

            "RMSE": np.sqrt(
                np.mean(errors ** 2)
            ),

            "WAPE": calculate_wape(
                actual,
                predicted
            ),

            "Forecast_Accuracy_KPI": (
                calculate_accuracy_kpi(
                    actual,
                    predicted
                )
            ),

            "Mean_Actual": np.mean(actual),

            "Mean_Predicted": np.mean(predicted)

        })

        records.append(record)

    return pd.DataFrame(records)


forecast_accuracy = calculate_forecast_metrics(
    test_df,
    ["Model", "Horizon"]
)

print("\nForecast accuracy KPI by model and horizon:")

display(
    forecast_accuracy.sort_values(
        "Forecast_Accuracy_KPI",
        ascending=False
    )
)


# ------------------------------------------------------------
# DEFINE HISTORICAL SURGE EVENTS AND SURGE THRESHOLDS
# ------------------------------------------------------------

# Use only the historical training period to determine
# the surge threshold, avoiding testing-data leakage.

historical_sorted = historical_df.sort_values(
    DATE_COLUMN
).reset_index(drop=True)

split_index = int(
    len(historical_sorted) * (1 - TEST_SIZE)
)

training_history = historical_sorted.iloc[
    :split_index
].copy()

testing_history = historical_sorted.iloc[
    split_index:
].copy()

surge_threshold = training_history[
    TARGET_COLUMN
].quantile(SURGE_QUANTILE)

historical_df["Surge_Threshold"] = surge_threshold

historical_df["Actual_Surge"] = (
    historical_df[TARGET_COLUMN] >= surge_threshold
)

historical_df["Actual_Regime"] = np.where(
    historical_df["Actual_Surge"],
    "Surge",
    "Stable"
)

print("\nSurge definition:")
print(
    f"Demand >= training-period {SURGE_QUANTILE:.0%} "
    "quantile"
)

print(
    "Surge threshold:",
    round(surge_threshold, 2)
)

print(
    "Historical surge events:",
    int(historical_df["Actual_Surge"].sum())
)

print(
    "Historical stable observations:",
    int((~historical_df["Actual_Surge"]).sum())
)


# ------------------------------------------------------------
# ASSIGN HISTORICAL REGIMES TO FORECAST PREDICTIONS
# ------------------------------------------------------------

actual_lookup = historical_df[
    [DATE_COLUMN, TARGET_COLUMN, "Actual_Surge"]
].copy()

actual_lookup = actual_lookup.rename(
    columns={
        DATE_COLUMN: "Lookup_Date",
        TARGET_COLUMN: "Historical_Actual",
        "Actual_Surge": "Historical_Surge"
    }
)

test_df["Predicted_Surge"] = (
    test_df["Predicted"] >= surge_threshold
)

if test_df["Actual_Target_Date"].notna().any():

    test_df = test_df.merge(
        actual_lookup,
        left_on="Actual_Target_Date",
        right_on="Lookup_Date",
        how="left"
    )

    test_df["Actual_Surge"] = (
        test_df["Historical_Surge"]
        .fillna(test_df["Actual"] >= surge_threshold)
        .astype(bool)
    )

    test_df = test_df.drop(
        columns=[
            "Lookup_Date",
            "Historical_Actual",
            "Historical_Surge"
        ],
        errors="ignore"
    )

else:

    test_df["Actual_Surge"] = (
        test_df["Actual"] >= surge_threshold
    )

test_df["Actual_Regime"] = np.where(
    test_df["Actual_Surge"],
    "Surge",
    "Stable"
)


# ------------------------------------------------------------
# MEASURE SURGE WARNING PRECISION AND RECALL
# ------------------------------------------------------------

def calculate_surge_metrics(df):

    records = []

    for (model, horizon), group in df.groupby(
        ["Model", "Horizon"]
    ):

        actual = group["Actual_Surge"].astype(bool)

        predicted = group["Predicted_Surge"].astype(bool)

        true_positive = int(
            (actual & predicted).sum()
        )

        false_positive = int(
            ((~actual) & predicted).sum()
        )

        false_negative = int(
            (actual & (~predicted)).sum()
        )

        true_negative = int(
            ((~actual) & (~predicted)).sum()
        )

        precision = (
            true_positive /
            (true_positive + false_positive)
            if true_positive + false_positive > 0
            else np.nan
        )

        recall = (
            true_positive /
            (true_positive + false_negative)
            if true_positive + false_negative > 0
            else np.nan
        )

        if (
            pd.notna(precision) and
            pd.notna(recall) and
            precision + recall > 0
        ):

            f1 = (
                2 * precision * recall /
                (precision + recall)
            )

        else:
            f1 = np.nan

        records.append({

            "Model": model,

            "Horizon": horizon,

            "Sample_Size": len(group),

            "True_Positive": true_positive,

            "False_Positive": false_positive,

            "False_Negative": false_negative,

            "True_Negative": true_negative,

            "Surge_Precision": precision,

            "Surge_Recall": recall,

            "Surge_F1": f1

        })

    return pd.DataFrame(records)


surge_metrics = calculate_surge_metrics(
    test_df
)

print("\nSurge warning precision and recall:")

display(
    surge_metrics.sort_values(
        "Surge_F1",
        ascending=False
    )
)


# ------------------------------------------------------------
# CALCULATE SURGE WARNING LEAD TIME
# ------------------------------------------------------------

# Lead time is measured in calendar days between the first
# warning and the start of an observed surge episode.
#
# Only forecasts with a known forecast origin are eligible.

def calculate_surge_lead_time(df):

    if not df["Forecast_Origin"].notna().any():

        return pd.DataFrame(
            columns=[
                "Model",
                "Horizon",
                "Surge_Events",
                "Detected_Events",
                "Mean_Lead_Time_Days",
                "Median_Lead_Time_Days"
            ]
        )

    records = []

    for (model, horizon), group in df.groupby(
        ["Model", "Horizon"]
    ):

        group = group.dropna(
            subset=[
                "Forecast_Origin",
                "Actual_Target_Date"
            ]
        ).copy()

        if group.empty:
            continue

        group["Predicted_Surge"] = (
            group["Predicted"] >= surge_threshold
        )

        group = group.sort_values(
            "Actual_Target_Date"
        )

        surge_dates = (
            group.loc[
                group["Actual_Surge"],
                "Actual_Target_Date"
            ]
            .dropna()
            .sort_values()
            .drop_duplicates()
        )

        lead_times = []

        for event_date in surge_dates:

            warnings_before_event = group[
                (group["Predicted_Surge"]) &
                (group["Forecast_Origin"] < event_date) &
                (
                    group["Forecast_Origin"]
                    >= event_date - pd.Timedelta(days=int(horizon))
                )
            ]

            if warnings_before_event.empty:
                continue

            earliest_warning = (
                warnings_before_event["Forecast_Origin"].min()
            )

            lead_time = (
                event_date - earliest_warning
            ).total_seconds() / 86400

            lead_times.append(lead_time)

        records.append({

            "Model": model,

            "Horizon": horizon,

            "Surge_Events": len(surge_dates),

            "Detected_Events": len(lead_times),

            "Mean_Lead_Time_Days": (
                np.mean(lead_times)
                if lead_times else np.nan
            ),

            "Median_Lead_Time_Days": (
                np.median(lead_times)
                if lead_times else np.nan
            )

        })

    return pd.DataFrame(records)


surge_lead_time = calculate_surge_lead_time(
    test_df
)

print("\nSurge warning lead-time analysis:")

if surge_lead_time.empty:

    print(
        "Lead time unavailable: forecast-origin dates "
        "are not available in the prediction records."
    )

else:

    display(surge_lead_time)


# ------------------------------------------------------------
# LOAD RELIABLE OPERATIONAL CAPACITY INPUT
# ------------------------------------------------------------

# Capacity can be provided through:
# 1. CAPACITY_THRESHOLD variable above.
# 2. outputs/capacity_threshold.csv
# 3. data/operational/capacity_threshold.csv
#
# The CSV must contain a column named capacity_threshold.
#
# This value must represent a verified operational capacity,
# not a statistically estimated demand threshold.

capacity_source = None

capacity_candidates = [
    OUTPUT_DIR / "capacity_threshold.csv",

    PROJECT_ROOT /
    "data" /
    "operational" /
    "capacity_threshold.csv"
]

if CAPACITY_THRESHOLD is not None:

    capacity_source = "User-provided configuration"

else:

    for capacity_file in capacity_candidates:

        if capacity_file.exists():

            capacity_data = pd.read_csv(
                capacity_file
            )

            capacity_data.columns = [
                str(col).strip().lower()
                for col in capacity_data.columns
            ]

            if "capacity_threshold" in capacity_data.columns:

                values = pd.to_numeric(
                    capacity_data["capacity_threshold"],
                    errors="coerce"
                ).dropna()

                if not values.empty:

                    CAPACITY_THRESHOLD = float(
                        values.iloc[-1]
                    )

                    capacity_source = str(
                        capacity_file
                    )

                    break

if CAPACITY_THRESHOLD is not None:

    CAPACITY_THRESHOLD = float(
        CAPACITY_THRESHOLD
    )

    if (
        not np.isfinite(CAPACITY_THRESHOLD)
        or CAPACITY_THRESHOLD <= 0
    ):

        raise ValueError(
            "Operational capacity must be a positive finite value."
        )

    print("\nVerified capacity threshold:", CAPACITY_THRESHOLD)
    print("Capacity source:", capacity_source)

else:

    print("\nCAPACITY THRESHOLD NOT AVAILABLE")

    print(
        "Capacity utilization and breach probability "
        "will be marked unavailable."
    )


# ------------------------------------------------------------
# CALCULATE CAPACITY UTILIZATION
# ------------------------------------------------------------

if CAPACITY_THRESHOLD is not None:

    historical_df["Capacity_Utilization"] = (
        historical_df[TARGET_COLUMN] /
        CAPACITY_THRESHOLD
    ) * 100

    historical_df["Capacity_Breach"] = (
        historical_df[TARGET_COLUMN] >
        CAPACITY_THRESHOLD
    )

    test_df["Capacity_Utilization"] = (
        test_df["Predicted"] /
        CAPACITY_THRESHOLD
    ) * 100

    test_df["Predicted_Capacity_Breach"] = (
        test_df["Predicted"] >
        CAPACITY_THRESHOLD
    )

    capacity_metrics = (
        historical_df.groupby("Actual_Regime")
        .agg(
            Mean_Capacity_Utilization=(
                "Capacity_Utilization", "mean"
            ),

            Maximum_Capacity_Utilization=(
                "Capacity_Utilization", "max"
            ),

            Historical_Breaches=(
                "Capacity_Breach", "sum"
            ),

            Observations=(
                "Capacity_Breach", "count"
            )
        )
        .reset_index()
    )

    print("\nHistorical capacity utilization:")

    display(capacity_metrics)

else:

    historical_df["Capacity_Utilization"] = np.nan
    historical_df["Capacity_Breach"] = pd.NA

    test_df["Capacity_Utilization"] = np.nan

    test_df["Predicted_Capacity_Breach"] = pd.NA

    capacity_metrics = pd.DataFrame(
        columns=[
            "Actual_Regime",
            "Mean_Capacity_Utilization",
            "Maximum_Capacity_Utilization",
            "Historical_Breaches",
            "Observations"
        ]
    )


# ------------------------------------------------------------
# CALCULATE CAPACITY BREACH PROBABILITY
# ------------------------------------------------------------

# Use validation residuals to estimate the probability that
# actual demand exceeds the verified operational capacity.

if CAPACITY_THRESHOLD is not None:

    validation_residuals = (
        validation_df["Actual"] -
        validation_df["Predicted"]
    ).dropna().to_numpy()

    if len(validation_residuals) < 5:

        raise ValueError(
            "Insufficient validation residuals for "
            "capacity breach probability estimation."
        )

    rng = np.random.default_rng(RANDOM_STATE)

    breach_records = []

    for (model, horizon), group in test_df.groupby(
        ["Model", "Horizon"]
    ):

        predicted_values = group[
            "Predicted"
        ].to_numpy()

        simulated_values = (
            predicted_values[:, None] +
            rng.choice(
                validation_residuals,
                size=(
                    len(predicted_values),
                    1000
                ),
                replace=True
            )
        )

        simulated_values = np.maximum(
            simulated_values,
            0
        )

        breach_probabilities = (
            simulated_values > CAPACITY_THRESHOLD
        ).mean(axis=1)

        breach_records.append({

            "Model": model,

            "Horizon": horizon,

            "Sample_Size": len(group),

            "Capacity_Threshold": CAPACITY_THRESHOLD,

            "Mean_Breach_Probability": (
                breach_probabilities.mean()
            ),

            "Maximum_Breach_Probability": (
                breach_probabilities.max()
            ),

            "Predicted_Breach_Count": int(
                (
                    predicted_values >
                    CAPACITY_THRESHOLD
                ).sum()
            )

        })

    capacity_breach_metrics = pd.DataFrame(
        breach_records
    )

    print("\nCapacity breach probability:")

    display(capacity_breach_metrics)

else:

    capacity_breach_metrics = pd.DataFrame(
        columns=[
            "Model",
            "Horizon",
            "Sample_Size",
            "Capacity_Threshold",
            "Mean_Breach_Probability",
            "Maximum_Breach_Probability",
            "Predicted_Breach_Count"
        ]
    )


# ------------------------------------------------------------
# CALCULATE FORECAST REVISIONS
# ------------------------------------------------------------

# Revisions compare forecasts for the same target date,
# model and horizon across different forecast origins.

revision_columns = [
    "Model",
    "Horizon",
    "Actual_Target_Date",
    "Forecast_Origin"
]

if (
    test_df["Forecast_Origin"].notna().any()
    and test_df["Actual_Target_Date"].notna().any()
):

    revision_data = test_df.dropna(
        subset=[
            "Model",
            "Horizon",
            "Actual_Target_Date",
            "Forecast_Origin"
        ]
    ).copy()

    revision_data = revision_data.sort_values(
        [
            "Model",
            "Horizon",
            "Actual_Target_Date",
            "Forecast_Origin"
        ]
    )

    revision_data["Previous_Forecast"] = (
        revision_data.groupby(
            [
                "Model",
                "Horizon",
                "Actual_Target_Date"
            ]
        )["Predicted"].shift(1)
    )

    revision_data["Forecast_Revision"] = (
        revision_data["Predicted"] -
        revision_data["Previous_Forecast"]
    )

    revision_data["Absolute_Forecast_Revision"] = (
        revision_data["Forecast_Revision"].abs()
    )

    revision_metrics = (
        revision_data.dropna(
            subset=["Forecast_Revision"]
        )
        .groupby(["Model", "Horizon"])
        .agg(
            Revision_Count=(
                "Forecast_Revision", "count"
            ),

            Mean_Absolute_Revision=(
                "Absolute_Forecast_Revision", "mean"
            ),

            Mean_Signed_Revision=(
                "Forecast_Revision", "mean"
            ),

            Maximum_Absolute_Revision=(
                "Absolute_Forecast_Revision", "max"
            )
        )
        .reset_index()
    )

else:

    revision_data = pd.DataFrame()

    revision_metrics = pd.DataFrame(
        columns=[
            "Model",
            "Horizon",
            "Revision_Count",
            "Mean_Absolute_Revision",
            "Mean_Signed_Revision",
            "Maximum_Absolute_Revision"
        ]
    )

    print(
        "\nForecast revisions unavailable: "
        "forecast-origin dates are required."
    )

print("\nForecast revision metrics:")

display(revision_metrics)


# ------------------------------------------------------------
# CALCULATE FORECAST STABILITY KPI
# ------------------------------------------------------------

# Definition:
# Stability KPI = 100 * (1 - normalized mean absolute revision)
#
# Normalization uses mean actual demand.
# A higher score indicates smaller revisions.

if not revision_metrics.empty:

    actual_scale = max(
        float(test_df["Actual"].abs().mean()),
        1.0
    )

    revision_metrics["Normalized_Revision"] = (
        revision_metrics["Mean_Absolute_Revision"] /
        actual_scale
    )

    revision_metrics["Forecast_Stability_KPI"] = (
        100 * (
            1 - revision_metrics[
                "Normalized_Revision"
            ]
        ).clip(0, 1)
    )

else:

    revision_metrics["Normalized_Revision"] = np.nan

    revision_metrics["Forecast_Stability_KPI"] = np.nan


# ------------------------------------------------------------
# TEST ALTERNATIVE TRAINING WINDOWS
# ------------------------------------------------------------

# Sensitivity experiment:
# Compare a rolling mean forecasting benchmark across
# alternative chronological training windows.
#
# This is a controlled sensitivity benchmark, not a claim
# that the production models were retrained.

sensitivity_series = (
    historical_df.set_index(DATE_COLUMN)[TARGET_COLUMN]
    .sort_index()
)

sensitivity_records = []

for window_fraction in TRAINING_WINDOWS:

    split = int(
        len(sensitivity_series) *
        (1 - TEST_SIZE)
    )

    train_values = sensitivity_series.iloc[:split].copy()

    test_values = sensitivity_series.iloc[split:].copy()

    window_size = max(
        2,
        int(len(train_values) * window_fraction)
    )

    if len(train_values) < window_size:
        continue

    for missing_method in [
        "drop",
        "interpolate",
        "forward_fill"
    ]:

        if missing_method == "drop":

            processed = sensitivity_series.dropna()

        elif missing_method == "interpolate":

            processed = sensitivity_series.interpolate(
                method="time",
                limit_area="inside"
            )

        else:

            processed = sensitivity_series.ffill()

        processed_train = processed.iloc[:split]

        processed_test = processed.iloc[split:]

        if (
            len(processed_train) <= window_size
            or len(processed_test) == 0
        ):
            continue

        # Rolling mean benchmark.

        historical_values = list(
            processed_train.dropna().to_numpy()
        )

        actual_values = []
        predicted_values = []

        for actual_value in processed_test:

            if not historical_values or pd.isna(actual_value):
                continue

            recent_values = historical_values[
                -window_size:
            ]

            prediction = np.mean(recent_values)

            actual_values.append(actual_value)

            predicted_values.append(prediction)

            historical_values.append(actual_value)

        if len(actual_values) == 0:
            continue

        sensitivity_records.append({

            "Training_Window": window_fraction,

            "Missing_Data_Treatment": missing_method,

            "Training_Observations": len(
                processed_train.dropna()
            ),

            "Testing_Observations": len(actual_values),

            "MAE": np.mean(
                np.abs(
                    np.array(actual_values) -
                    np.array(predicted_values)
                )
            ),

            "RMSE": np.sqrt(
                np.mean(
                    (
                        np.array(actual_values) -
                        np.array(predicted_values)
                    ) ** 2
                )
            ),

            "WAPE": calculate_wape(
                actual_values,
                predicted_values
            )

        })


training_sensitivity = pd.DataFrame(
    sensitivity_records
)

print("\nAlternative training-window sensitivity:")

display(training_sensitivity)


# ------------------------------------------------------------
# TEST ALTERNATIVE MISSING-DATA TREATMENTS
# ------------------------------------------------------------

if not training_sensitivity.empty:

    missing_data_sensitivity = (
        training_sensitivity.groupby(
            "Missing_Data_Treatment"
        )
        .agg(
            Mean_MAE=("MAE", "mean"),

            Mean_RMSE=("RMSE", "mean"),

            Mean_WAPE=("WAPE", "mean"),

            Experiment_Count=("WAPE", "count")
        )
        .reset_index()
    )

else:

    missing_data_sensitivity = pd.DataFrame()

print("\nMissing-data treatment sensitivity:")

display(missing_data_sensitivity)


# ------------------------------------------------------------
# TEST MODEL PERFORMANCE ACROSS OPERATIONAL REGIMES
# ------------------------------------------------------------

regime_records = []

for (model, horizon, regime), group in test_df.groupby(
    ["Model", "Horizon", "Actual_Regime"]
):

    regime_records.append({

        "Model": model,

        "Horizon": horizon,

        "Regime": regime,

        "Sample_Size": len(group),

        "MAE": np.mean(
            np.abs(
                group["Actual"] -
                group["Predicted"]
            )
        ),

        "RMSE": np.sqrt(
            np.mean(
                (
                    group["Actual"] -
                    group["Predicted"]
                ) ** 2
            )
        ),

        "WAPE": calculate_wape(
            group["Actual"],
            group["Predicted"]
        )

    })

regime_performance = pd.DataFrame(
    regime_records
)

print("\nModel performance across surge and stable regimes:")

display(regime_performance)


# ------------------------------------------------------------
# CALCULATE MODEL ROBUSTNESS KPI
# ------------------------------------------------------------

# Definition:
# Robustness KPI measures consistency of WAPE across
# available surge/stable regimes.
#
# Lower variation in WAPE means greater regime robustness.
# A model with insufficient regime samples receives NaN.

robustness_records = []

for model, group in regime_performance.groupby("Model"):

    regime_wape = (
        group.groupby("Regime")["WAPE"]
        .mean()
        .dropna()
    )

    if len(regime_wape) >= 2:

        mean_wape = regime_wape.mean()

        variation = (
            regime_wape.std(ddof=0) /
            max(mean_wape, 1e-9)
        )

        robustness = (
            100 * (1 - variation)
        )

        robustness = np.clip(
            robustness,
            0,
            100
        )

    else:

        variation = np.nan
        robustness = np.nan

    robustness_records.append({

        "Model": model,

        "Regimes_Evaluated": len(regime_wape),

        "Mean_Regime_WAPE": (
            regime_wape.mean()
            if len(regime_wape) else np.nan
        ),

        "Regime_WAPE_Variation": variation,

        "Model_Robustness_KPI": robustness

    })

robustness_metrics = pd.DataFrame(
    robustness_records
)

print("\nModel robustness KPI:")

display(
    robustness_metrics.sort_values(
        "Model_Robustness_KPI",
        ascending=False
    )
)


# ------------------------------------------------------------
# GENERATE CONSOLIDATED FIVE-KPI SCORECARD
# ------------------------------------------------------------

# All five KPIs use a 0-100 scale where possible.
#
# Accuracy:
# 100 * (1 - WAPE)
#
# Surge Detection:
# 100 * Surge F1
#
# Capacity Risk:
# 100 * (1 - breach probability)
#
# Forecast Stability:
# Normalized forecast revision score.
#
# Model Robustness:
# Consistency of WAPE across surge and stable regimes.
#
# Missing operational data is NOT replaced with invented values.

scorecard_records = []

for model in sorted(test_df["Model"].unique()):

    model_accuracy = forecast_accuracy[
        forecast_accuracy["Model"] == model
    ]

    model_surge = surge_metrics[
        surge_metrics["Model"] == model
    ]

    model_capacity = capacity_breach_metrics[
        capacity_breach_metrics["Model"] == model
    ]

    model_stability = revision_metrics[
        revision_metrics["Model"] == model
    ]

    model_robustness = robustness_metrics[
        robustness_metrics["Model"] == model
    ]

    accuracy_value = (
        model_accuracy["Forecast_Accuracy_KPI"].mean()
        if not model_accuracy.empty
        else np.nan
    )

    surge_value = (
        model_surge["Surge_F1"].mean() * 100
        if model_surge["Surge_F1"].notna().any()
        else np.nan
    )

    capacity_value = (
        (
            1 - model_capacity[
                "Mean_Breach_Probability"
            ].mean()
        ) * 100
        if not model_capacity.empty
        else np.nan
    )

    stability_value = (
        model_stability[
            "Forecast_Stability_KPI"
        ].mean()
        if (
            not model_stability.empty
            and model_stability[
                "Forecast_Stability_KPI"
            ].notna().any()
        )
        else np.nan
    )

    robustness_value = (
        model_robustness[
            "Model_Robustness_KPI"
        ].iloc[0]
        if (
            not model_robustness.empty
            and pd.notna(
                model_robustness[
                    "Model_Robustness_KPI"
                ].iloc[0]
            )
        )
        else np.nan
    )

    values = [
        accuracy_value,
        surge_value,
        capacity_value,
        stability_value,
        robustness_value
    ]

    available_values = [
        value for value in values
        if pd.notna(value)
    ]

    overall_score = (
        np.mean(available_values)
        if available_values
        else np.nan
    )

    scorecard_records.append({

        "Model": model,

        "Forecast_Accuracy_KPI": accuracy_value,

        "Surge_Detection_KPI": surge_value,

        "Capacity_Risk_KPI": capacity_value,

        "Forecast_Stability_KPI": stability_value,

        "Model_Robustness_KPI": robustness_value,

        "Available_KPIs": len(available_values),

        "Overall_Available_KPI_Score": overall_score

    })

five_kpi_scorecard = pd.DataFrame(
    scorecard_records
)

five_kpi_scorecard = five_kpi_scorecard.sort_values(
    "Overall_Available_KPI_Score",
    ascending=False,
    na_position="last"
).reset_index(drop=True)

print("\nCONSOLIDATED FIVE-KPI SCORECARD")

display(five_kpi_scorecard)


# ------------------------------------------------------------
# DOCUMENT KPI INTERPRETATION AND LIMITATIONS
# ------------------------------------------------------------

kpi_documentation = """
OPERATIONAL KPIs AND CAPACITY RISK ANALYSIS
============================================

1. FORECAST ACCURACY KPI

Definition:
    Forecast Accuracy = max(0, 1 - WAPE) * 100

WAPE measures total absolute forecasting error relative
to total observed demand.

Higher values indicate better aggregate forecast accuracy.

2. HISTORICAL SURGE EVENTS

Definition:
    Surge = Actual demand >= 90th percentile of training demand.

The threshold is estimated from historical training data
rather than testing data.

3. SURGE WARNING PRECISION

Definition:
    True Positive / (True Positive + False Positive)

Higher precision means fewer false surge warnings.

4. SURGE WARNING RECALL

Definition:
    True Positive / (True Positive + False Negative)

Higher recall means more actual surge periods are detected.

5. SURGE WARNING LEAD TIME

Definition:
    Calendar days between an eligible forecast warning
    and the start of an observed surge event.

Requires forecast-origin dates.

6. OPERATIONAL CAPACITY THRESHOLD

The threshold must come from an authoritative operational
capacity source.

A historical demand quantile is not a substitute for
actual HHS operational capacity.

7. CAPACITY UTILIZATION

Definition:
    Actual or forecast demand / Operational capacity * 100

Values above 100% indicate demand exceeding the supplied
capacity threshold.

8. CAPACITY BREACH PROBABILITY

Estimated using validation residual bootstrap simulations.

Definition:
    Proportion of simulated future demand values exceeding
    the verified operational capacity threshold.

The estimate depends on the validation residual distribution.

9. FORECAST REVISIONS

Definition:
    Difference between successive forecasts for the same
    target date, model and horizon.

Requires multiple forecast origins.

10. FORECAST STABILITY KPI

Definition:
    100 * (1 - normalized mean absolute forecast revision)

Higher values indicate smaller forecast revisions.

11. TRAINING WINDOW SENSITIVITY

Alternative chronological training windows are compared
using a rolling-mean forecasting benchmark.

This is a sensitivity experiment, not a retraining claim
for every production forecasting model.

12. MISSING-DATA TREATMENT SENSITIVITY

Compares:
- Dropping missing observations.
- Time-based interpolation.
- Forward filling.

These methods are sensitivity alternatives, not automatic
replacements for the primary cleaned dataset.

13. PERFORMANCE ACROSS REGIMES

Model performance is measured separately during:
- Surge periods.
- Stable periods.

14. MODEL ROBUSTNESS KPI

Definition:
    100 * (1 - coefficient of variation of regime WAPE)

The score is restricted to 0-100.

It is available only when both surge and stable regimes
have sufficient evaluated observations.

15. FIVE-KPI SCORECARD

The five indicators are:
- Forecast accuracy.
- Surge detection.
- Capacity risk.
- Forecast stability.
- Model robustness.

The overall available KPI score is the unweighted mean
of available indicators.

Missing indicators are excluded rather than assigned
artificial scores.

LIMITATIONS
===========

1. Historical surge thresholds do not represent an
   officially approved operational surge definition.

2. Forecast warning precision and recall depend on the
   selected threshold and forecast horizon.

3. Lead time cannot be reliably calculated without
   forecast-origin dates.

4. Capacity utilization and breach probability cannot
   be established without verified operational capacity.

5. Bootstrap breach probabilities assume validation
   residuals provide a reasonable representation of
   future forecasting errors.

6. Structural changes and unprecedented demand surges
   may invalidate historical residual distributions.

7. Forecast revisions require repeated forecasts for
   identical target dates.

8. Stability scores are not direct measures of forecast
   accuracy.

9. Training-window and missing-data sensitivity results
   use a rolling-mean benchmark rather than retraining
   all production models.

10. Regime robustness is dependent on the number and
    distribution of surge and stable observations.

11. The overall KPI score is a descriptive composite,
    not an official HHS operational performance rating.

12. Operational decisions should consider data quality,
    staffing, facility availability and other constraints
    that are not represented by the forecasting model.
"""

with open(
    KPI_DIR / "kpi_interpretation_and_limitations.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write(kpi_documentation)


# ------------------------------------------------------------
# SAVE ALL KPI TABLES
# ------------------------------------------------------------

output_tables = {

    "forecast_accuracy_kpi.csv":
        forecast_accuracy,

    "historical_surge_events.csv":
        historical_df,

    "surge_warning_metrics.csv":
        surge_metrics,

    "surge_lead_time.csv":
        surge_lead_time,

    "capacity_utilization.csv":
        capacity_metrics,

    "capacity_breach_probability.csv":
        capacity_breach_metrics,

    "forecast_revisions.csv":
        revision_data,

    "forecast_stability_kpi.csv":
        revision_metrics,

    "training_window_sensitivity.csv":
        training_sensitivity,

    "missing_data_sensitivity.csv":
        missing_data_sensitivity,

    "model_regime_performance.csv":
        regime_performance,

    "model_robustness_kpi.csv":
        robustness_metrics,

    "five_kpi_scorecard.csv":
        five_kpi_scorecard

}

for filename, dataframe in output_tables.items():

    dataframe.to_csv(
        KPI_DIR / filename,
        index=False
    )


# ------------------------------------------------------------
# VISUALIZATION: FORECAST ACCURACY
# ------------------------------------------------------------

accuracy_plot = (
    forecast_accuracy.groupby("Model")
    ["Forecast_Accuracy_KPI"]
    .mean()
    .dropna()
    .sort_values(ascending=False)
)

if not accuracy_plot.empty:

    fig, ax = plt.subplots(figsize=(12, 6))

    accuracy_plot.plot(
        kind="bar",
        ax=ax
    )

    ax.set_title("Forecast Accuracy KPI by Model")
    ax.set_xlabel("Forecasting Model")
    ax.set_ylabel("Forecast Accuracy (%)")

    ax.tick_params(
        axis="x",
        rotation=45
    )

    ax.grid(axis="y", alpha=0.25)

    plt.tight_layout()

    plt.savefig(
        KPI_DIR / "forecast_accuracy_kpi.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# VISUALIZATION: SURGE DETECTION
# ------------------------------------------------------------

surge_plot = (
    surge_metrics.groupby("Model")
    [["Surge_Precision", "Surge_Recall"]]
    .mean()
    .dropna(how="all")
)

if not surge_plot.empty:

    ax = surge_plot.plot(
        kind="bar",
        figsize=(12, 6)
    )

    ax.set_title("Surge Warning Precision and Recall")
    ax.set_xlabel("Forecasting Model")
    ax.set_ylabel("Score")

    ax.legend(
        ["Precision", "Recall"]
    )

    plt.xticks(rotation=45)
    plt.tight_layout()

    plt.savefig(
        KPI_DIR / "surge_detection_performance.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# VISUALIZATION: CAPACITY UTILIZATION
# ------------------------------------------------------------

if CAPACITY_THRESHOLD is not None:

    fig, ax = plt.subplots(figsize=(14, 5))

    ax.plot(
        historical_df[DATE_COLUMN],
        historical_df["Capacity_Utilization"],
        label="Historical Capacity Utilization"
    )

    ax.axhline(
        100,
        linestyle="--",
        label="Capacity Threshold"
    )

    ax.set_title("Historical HHS Care Capacity Utilization")
    ax.set_xlabel("Date")
    ax.set_ylabel("Capacity Utilization (%)")

    ax.legend()
    ax.grid(alpha=0.25)

    plt.tight_layout()

    plt.savefig(
        KPI_DIR / "capacity_utilization.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# VISUALIZATION: FIVE-KPI SCORECARD
# ------------------------------------------------------------

scorecard_plot = five_kpi_scorecard.set_index(
    "Model"
)[
    [
        "Forecast_Accuracy_KPI",
        "Surge_Detection_KPI",
        "Capacity_Risk_KPI",
        "Forecast_Stability_KPI",
        "Model_Robustness_KPI"
    ]
]

if not scorecard_plot.empty:

    fig, ax = plt.subplots(figsize=(15, 7))

    scorecard_plot.plot(
        kind="bar",
        ax=ax
    )

    ax.set_title("Consolidated Five-KPI Model Scorecard")
    ax.set_xlabel("Forecasting Model")
    ax.set_ylabel("KPI Score (0-100)")

    ax.set_ylim(0, 105)

    ax.legend(
        loc="upper left",
        bbox_to_anchor=(1, 1)
    )

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.savefig(
        KPI_DIR / "five_kpi_scorecard.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# FINAL EXECUTION SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("OPERATIONAL KPI AND CAPACITY RISK ANALYSIS COMPLETED")
print("=" * 80)

print("\nHistorical observations:", len(historical_df))

print("Validation predictions:", len(validation_df))

print("Testing predictions:", len(test_df))

print(
    "Models evaluated:",
    test_df["Model"].nunique()
)

print(
    "Surge threshold:",
    round(surge_threshold, 2)
)

print(
    "Historical surge observations:",
    int(historical_df["Actual_Surge"].sum())
)

print(
    "Capacity threshold:",
    CAPACITY_THRESHOLD
)

if CAPACITY_THRESHOLD is None:

    print(
        "Capacity risk status: Awaiting verified "
        "operational capacity input."
    )

else:

    print(
        "Capacity risk status: Calculated using "
        "the supplied capacity threshold."
    )

print("\nFive-KPI scorecard:")

display(five_kpi_scorecard)

print("\nGenerated output files:")

for file_path in sorted(KPI_DIR.iterdir()):

    if file_path.is_file():

        print(" -", file_path.name)

print("\nOutput directory:")
print(KPI_DIR)

print("\nAll available operational KPI calculations completed.")

display(
    Markdown(
        "**Operational KPI and capacity risk analysis completed.**"
    )
)
```
