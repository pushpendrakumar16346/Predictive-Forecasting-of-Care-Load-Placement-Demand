```python

# Scenario Analysis and Operational Planning

import os
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, Markdown

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# PROJECT CONFIGURATION
# ------------------------------------------------------------

RANDOM_STATE = 42
rng = np.random.default_rng(RANDOM_STATE)

TARGET_COLUMN = "children_in_hhs_care"
DATE_COLUMN = "date"

FORECAST_HORIZONS = [1, 7, 14, 30]

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = (
    PROJECT_ROOT /
    "data" /
    "processed" /
    "cleaned_care_data.csv"
)

OUTPUT_DIR = PROJECT_ROOT / "outputs"

SCENARIO_DIR = (
    OUTPUT_DIR / "scenario_analysis"
)

SCENARIO_DIR.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# CONFIGURABLE SCENARIO ASSUMPTIONS
# ------------------------------------------------------------

# Multipliers are explicit planning assumptions, not
# statistically established probabilities.

SCENARIO_ASSUMPTIONS = {

    "Low_Demand": {
        "inflow_multiplier": 0.80,
        "transfer_multiplier": 0.90,
        "discharge_multiplier": 1.10
    },

    "Baseline": {
        "inflow_multiplier": 1.00,
        "transfer_multiplier": 1.00,
        "discharge_multiplier": 1.00
    },

    "High_Demand": {
        "inflow_multiplier": 1.20,
        "transfer_multiplier": 1.10,
        "discharge_multiplier": 0.90
    }

}

# Capacity must be obtained from an authoritative
# operational source. Do not invent this value.

CAPACITY_THRESHOLD = None

CAPACITY_FILE_CANDIDATES = [
    OUTPUT_DIR / "capacity_threshold.csv",

    PROJECT_ROOT /
    "data" /
    "operational" /
    "capacity_threshold.csv"
]

# Number of bootstrap simulations.

N_SIMULATIONS = 1000

# Historical window used for flow-rate estimation.

HISTORICAL_WINDOW_DAYS = 90


print("=" * 80)
print("SCENARIO ANALYSIS AND OPERATIONAL PLANNING")
print("=" * 80)

print("Project directory:", PROJECT_ROOT)
print("Dataset:", DATA_PATH)
print("Output directory:", SCENARIO_DIR)


# ------------------------------------------------------------
# LOAD HISTORICAL DATA
# ------------------------------------------------------------

if not DATA_PATH.exists():

    raise FileNotFoundError(
        f"Cleaned dataset not found: {DATA_PATH}"
    )

df = pd.read_csv(DATA_PATH)

df.columns = [
    str(col).strip().lower().replace(" ", "_")
    for col in df.columns
]

REQUIRED_COLUMNS = [
    "date",
    "children_apprehended_and_placed_in_cbp_custody",
    "children_in_cbp_custody",
    "children_transferred_out_of_cbp_custody",
    "children_in_hhs_care",
    "children_discharged_from_hhs_care"
]

missing_columns = [
    col for col in REQUIRED_COLUMNS
    if col not in df.columns
]

if missing_columns:

    raise KeyError(
        f"Missing required dataset columns: {missing_columns}"
    )

df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)

NUMERIC_COLUMNS = REQUIRED_COLUMNS[1:]

for col in NUMERIC_COLUMNS:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

df = (
    df.dropna(subset=["date"])
    .sort_values("date")
    .drop_duplicates(subset=["date"], keep="last")
    .reset_index(drop=True)
)

if len(df) < 30:

    raise ValueError(
        "Insufficient historical observations for scenario analysis."
    )

print("\nHistorical observations:", len(df))

print(
    "Historical date range:",
    df["date"].min(),
    "to",
    df["date"].max()
)

print("\nDataset columns:")
print(df.columns.tolist())


# ------------------------------------------------------------
# VALIDATE STOCK AND FLOW VARIABLES
# ------------------------------------------------------------

STOCK_COLUMNS = [
    "children_in_cbp_custody",
    "children_in_hhs_care"
]

FLOW_COLUMNS = [
    "children_apprehended_and_placed_in_cbp_custody",
    "children_transferred_out_of_cbp_custody",
    "children_discharged_from_hhs_care"
]

for col in STOCK_COLUMNS:

    if df[col].dropna().empty:

        raise ValueError(
            f"No valid historical stock observations: {col}"
        )

    if (df[col].dropna() < 0).any():

        raise ValueError(
            f"Negative population values detected in {col}."
        )

for col in FLOW_COLUMNS:

    if df[col].dropna().empty:

        raise ValueError(
            f"No valid historical flow observations: {col}"
        )

    if (df[col].dropna() < 0).any():

        raise ValueError(
            f"Negative flow values detected in {col}."
        )


# ------------------------------------------------------------
# ANALYZE HISTORICAL REPORTING FREQUENCY
# ------------------------------------------------------------

df["Days_Between_Reports"] = (
    df["date"].diff().dt.total_seconds() / 86400
)

reporting_intervals = (
    df["Days_Between_Reports"]
    .dropna()
)

print("\nHistorical reporting frequency:")

print(
    "Median days between reports:",
    reporting_intervals.median()
)

print(
    "Mean days between reports:",
    reporting_intervals.mean()
)

print(
    "Maximum reporting gap:",
    reporting_intervals.max()
)

print(
    "Missing calendar dates are not treated as zero-demand days."
)


# ------------------------------------------------------------
# ESTIMATE HISTORICAL DAILY FLOW RATES
# ------------------------------------------------------------

# Assumption:
# Flow observations represent counts accumulated during
# the interval between consecutive reporting dates.
#
# Dividing by elapsed calendar days converts observed
# interval flows into approximate daily flow rates.
#
# This assumption must be checked against the source's
# reporting definitions before operational use.

for col in FLOW_COLUMNS:

    df[col + "_Daily_Rate"] = (
        df[col] /
        df["Days_Between_Reports"]
    )

    df.loc[
        df["Days_Between_Reports"] <= 0,
        col + "_Daily_Rate"
    ] = np.nan


# ------------------------------------------------------------
# ESTIMATE BASELINE OPERATIONAL FLOW RATES
# ------------------------------------------------------------

latest_date = df["date"].max()

recent_df = df[
    df["date"] >= (
        latest_date -
        pd.Timedelta(days=HISTORICAL_WINDOW_DAYS)
    )
].copy()

if len(recent_df) < 5:

    recent_df = df.copy()

FLOW_RATE_COLUMNS = {
    "inflow": (
        "children_apprehended_and_placed_in_cbp_custody_Daily_Rate"
    ),

    "transfer": (
        "children_transferred_out_of_cbp_custody_Daily_Rate"
    ),

    "discharge": (
        "children_discharged_from_hhs_care_Daily_Rate"
    )
}

baseline_rates = {}

for name, col in FLOW_RATE_COLUMNS.items():

    valid = recent_df[col].dropna()

    if valid.empty:

        raise ValueError(
            f"Unable to estimate baseline {name} rate."
        )

    # Median reduces sensitivity to unusually large reporting periods.

    baseline_rates[name] = max(
        0.0,
        float(valid.median())
    )

print("\nEstimated baseline daily flow rates:")

for name, value in baseline_rates.items():

    print(
        name.capitalize(),
        ":",
        round(value, 2),
        "children per day"
    )


# ------------------------------------------------------------
# DEFINE LOW, BASELINE AND HIGH-DEMAND SCENARIOS
# ------------------------------------------------------------

scenario_assumptions_df = pd.DataFrame.from_dict(
    SCENARIO_ASSUMPTIONS,
    orient="index"
).reset_index()

scenario_assumptions_df = (
    scenario_assumptions_df.rename(
        columns={"index": "Scenario"}
    )
)

print("\nScenario assumptions:")

display(scenario_assumptions_df)


# ------------------------------------------------------------
# DEFINE INFLOW ASSUMPTIONS
# ------------------------------------------------------------

# Inflow represents children apprehended and placed in
# CBP custody during the reporting interval.

# Scenario-specific inflow is derived from the historical
# daily inflow rate and the scenario multiplier.

# ------------------------------------------------------------
# DEFINE TRANSFER ASSUMPTIONS
# ------------------------------------------------------------

# Transfers out of CBP custody are treated as inflows
# into the HHS care population for this stock-flow model.

# ------------------------------------------------------------
# DEFINE DISCHARGE ASSUMPTIONS
# ------------------------------------------------------------

# Discharges reduce the HHS care population.

# ------------------------------------------------------------
# DEFINE CAPACITY ASSUMPTIONS
# ------------------------------------------------------------

if CAPACITY_THRESHOLD is None:

    for capacity_file in CAPACITY_FILE_CANDIDATES:

        if capacity_file.exists():

            capacity_data = pd.read_csv(
                capacity_file
            )

            capacity_data.columns = [
                str(col).strip().lower()
                for col in capacity_data.columns
            ]

            if "capacity_threshold" in capacity_data.columns:

                values = pd.to_numeric(
                    capacity_data["capacity_threshold"],
                    errors="coerce"
                ).dropna()

                if not values.empty:

                    CAPACITY_THRESHOLD = float(
                        values.iloc[-1]
                    )

                    break

if CAPACITY_THRESHOLD is not None:

    CAPACITY_THRESHOLD = float(
        CAPACITY_THRESHOLD
    )

    if (
        not np.isfinite(CAPACITY_THRESHOLD)
        or CAPACITY_THRESHOLD <= 0
    ):

        raise ValueError(
            "Capacity threshold must be a positive finite value."
        )

    print(
        "\nOperational capacity threshold:",
        CAPACITY_THRESHOLD
    )

else:

    print(
        "\nOperational capacity not provided. "
        "Capacity breach probability will remain unavailable."
    )


# ------------------------------------------------------------
# DEFINE INITIAL POPULATION STOCKS
# ------------------------------------------------------------

latest_hhs = df.dropna(
    subset=["children_in_hhs_care"]
).iloc[-1]

latest_cbp = df.dropna(
    subset=["children_in_cbp_custody"]
).iloc[-1]

INITIAL_HHS_POPULATION = float(
    latest_hhs["children_in_hhs_care"]
)

INITIAL_CBP_POPULATION = float(
    latest_cbp["children_in_cbp_custody"]
)

print("\nInitial population stocks:")

print(
    "HHS care population:",
    INITIAL_HHS_POPULATION
)

print(
    "CBP custody population:",
    INITIAL_CBP_POPULATION
)

print(
    "Forecast start date:",
    latest_date
)


# ------------------------------------------------------------
# PREPARE HISTORICAL NET FLOW DISTRIBUTIONS
# ------------------------------------------------------------

# The observed net flows are used to characterize
# uncertainty around the baseline flow assumptions.

recent_valid = recent_df.dropna(
    subset=[
        FLOW_RATE_COLUMNS["inflow"],
        FLOW_RATE_COLUMNS["transfer"],
        FLOW_RATE_COLUMNS["discharge"]
    ]
).copy()

if len(recent_valid) < 5:

    recent_valid = df.dropna(
        subset=[
            FLOW_RATE_COLUMNS["inflow"],
            FLOW_RATE_COLUMNS["transfer"],
            FLOW_RATE_COLUMNS["discharge"]
        ]
    ).copy()

if len(recent_valid) < 5:

    raise ValueError(
        "Insufficient complete historical flow observations "
        "for scenario uncertainty simulation."
    )

flow_residuals = pd.DataFrame({

    "inflow": (
        recent_valid[
            FLOW_RATE_COLUMNS["inflow"]
        ] - baseline_rates["inflow"]
    ),

    "transfer": (
        recent_valid[
            FLOW_RATE_COLUMNS["transfer"]
        ] - baseline_rates["transfer"]
    ),

    "discharge": (
        recent_valid[
            FLOW_RATE_COLUMNS["discharge"]
        ] - baseline_rates["discharge"]
    )

}).dropna()

if len(flow_residuals) < 5:

    raise ValueError(
        "Insufficient residual observations for uncertainty simulation."
    )

print(
    "\nHistorical flow residual observations:",
    len(flow_residuals)
)


# ------------------------------------------------------------
# GENERATE SCENARIO-SPECIFIC NET FLOWS AND FORECASTS
# ------------------------------------------------------------

scenario_forecasts = []

simulation_records = []

for scenario_name, assumptions in SCENARIO_ASSUMPTIONS.items():

    print(
        "\nGenerating scenario:",
        scenario_name
    )

    hhs_population = INITIAL_HHS_POPULATION

    cbp_population = INITIAL_CBP_POPULATION

    scenario_daily_records = []

    for day in range(1, max(FORECAST_HORIZONS) + 1):

        # Scenario-specific flow assumptions.

        daily_inflow = (
            baseline_rates["inflow"] *
            assumptions["inflow_multiplier"]
        )

        daily_transfer = (
            baseline_rates["transfer"] *
            assumptions["transfer_multiplier"]
        )

        daily_discharge = (
            baseline_rates["discharge"] *
            assumptions["discharge_multiplier"]
        )

        # Calculate net population flows.

        cbp_net_flow = (
            daily_inflow - daily_transfer
        )

        hhs_net_flow = (
            daily_transfer - daily_discharge
        )

        # Update population stocks.

        cbp_population = max(
            0,
            cbp_population + cbp_net_flow
        )

        hhs_population = max(
            0,
            hhs_population + hhs_net_flow
        )

        forecast_date = (
            latest_date +
            pd.Timedelta(days=day)
        )

        record = {

            "Scenario": scenario_name,

            "Forecast_Date": forecast_date,

            "Horizon": day,

            "Daily_Inflows": daily_inflow,

            "Daily_Transfers": daily_transfer,

            "Daily_Discharges": daily_discharge,

            "CBP_Net_Flow": cbp_net_flow,

            "HHS_Net_Flow": hhs_net_flow,

            "Projected_CBP_Custody": cbp_population,

            "Projected_HHS_Care": hhs_population,

            "Projected_Total_Population": (
                cbp_population + hhs_population
            )

        }

        if CAPACITY_THRESHOLD is not None:

            record["Capacity_Utilization"] = (
                hhs_population /
                CAPACITY_THRESHOLD
            ) * 100

            record["Capacity_Breach"] = (
                hhs_population >
                CAPACITY_THRESHOLD
            )

        else:

            record["Capacity_Utilization"] = np.nan

            record["Capacity_Breach"] = pd.NA

        scenario_daily_records.append(record)

    scenario_daily_df = pd.DataFrame(
        scenario_daily_records
    )

    scenario_forecasts.append(
        scenario_daily_df
    )

    # Save daily scenario simulation for bootstrap.

    simulation_records.append({

        "Scenario": scenario_name,

        "Assumptions": assumptions,

        "Daily_Data": scenario_daily_df

    })


scenario_forecasts_df = pd.concat(
    scenario_forecasts,
    ignore_index=True
)

print("\nScenario-specific forecasts generated.")

display(
    scenario_forecasts_df.head(15)
)


# ------------------------------------------------------------
# GENERATE SCENARIO-SPECIFIC FORECAST SUMMARIES
# ------------------------------------------------------------

scenario_summary_records = []

for scenario_name, group in scenario_forecasts_df.groupby(
    "Scenario"
):

    for horizon in FORECAST_HORIZONS:

        selected = group[
            group["Horizon"] == horizon
        ]

        if selected.empty:
            continue

        row = selected.iloc[-1]

        scenario_summary_records.append({

            "Scenario": scenario_name,

            "Horizon": horizon,

            "Projected_HHS_Care": (
                row["Projected_HHS_Care"]
            ),

            "Projected_CBP_Custody": (
                row["Projected_CBP_Custody"]
            ),

            "Projected_Total_Population": (
                row["Projected_Total_Population"]
            ),

            "Daily_Inflows": (
                row["Daily_Inflows"]
            ),

            "Daily_Transfers": (
                row["Daily_Transfers"]
            ),

            "Daily_Discharges": (
                row["Daily_Discharges"]
            ),

            "HHS_Net_Flow": (
                row["HHS_Net_Flow"]
            ),

            "CBP_Net_Flow": (
                row["CBP_Net_Flow"]
            ),

            "Projected_Peak_Care_Load": (
                group[
                    group["Horizon"] <= horizon
                ]["Projected_HHS_Care"].max()
            ),

            "Capacity_Utilization": (
                row["Capacity_Utilization"]
            )

        })

scenario_summary = pd.DataFrame(
    scenario_summary_records
)

print("\nScenario-specific forecast summary:")

display(scenario_summary)


# ------------------------------------------------------------
# CALCULATE PROJECTED HHS CARE POPULATION
# ------------------------------------------------------------

hhs_scenario_comparison = scenario_summary[
    [
        "Scenario",
        "Horizon",
        "Projected_HHS_Care",
        "Projected_Peak_Care_Load",
        "HHS_Net_Flow"
    ]
].copy()

print("\nProjected HHS care population:")

display(hhs_scenario_comparison)


# ------------------------------------------------------------
# CALCULATE PROJECTED CBP CUSTODY POPULATION
# ------------------------------------------------------------

cbp_scenario_comparison = scenario_summary[
    [
        "Scenario",
        "Horizon",
        "Projected_CBP_Custody",
        "CBP_Net_Flow",
        "Daily_Inflows",
        "Daily_Transfers"
    ]
].copy()

print("\nProjected CBP custody population:")

display(cbp_scenario_comparison)


# ------------------------------------------------------------
# CALCULATE PROJECTED PEAK CARE LOAD
# ------------------------------------------------------------

peak_load_records = []

for (scenario_name, horizon), group in (
    scenario_forecasts_df.groupby(
        [
            "Scenario",
            scenario_forecasts_df["Horizon"].map(
                lambda x: min(
                    [h for h in FORECAST_HORIZONS if h >= x],
                    default=max(FORECAST_HORIZONS)
                )
            )
        ]
    )
):

    peak_load_records.append({

        "Scenario": scenario_name,

        "Horizon": horizon,

        "Peak_HHS_Care_Load": (
            group["Projected_HHS_Care"].max()
        )

    })

peak_load_summary = pd.DataFrame(
    peak_load_records
)

print("\nProjected peak HHS care load:")

display(peak_load_summary)


# ------------------------------------------------------------
# SIMULATE UNCERTAINTY ACROSS SCENARIOS
# ------------------------------------------------------------

# Residual bootstrap:
# Historical deviations from baseline flow rates are sampled
# to produce alternative possible future population paths.
#
# These are scenario-conditional simulations, not probabilities
# assigned to the low, baseline or high-demand scenarios.

uncertainty_records = []

simulation_path_records = []

for scenario_name, assumptions in SCENARIO_ASSUMPTIONS.items():

    for simulation_id in range(N_SIMULATIONS):

        sampled_indices = rng.integers(
            0,
            len(flow_residuals),
            size=max(FORECAST_HORIZONS)
        )

        sampled_errors = (
            flow_residuals.iloc[
                sampled_indices
            ].reset_index(drop=True)
        )

        hhs_sim = INITIAL_HHS_POPULATION

        cbp_sim = INITIAL_CBP_POPULATION

        for day in range(1, max(FORECAST_HORIZONS) + 1):

            inflow = max(
                0,
                baseline_rates["inflow"] *
                assumptions["inflow_multiplier"] +
                sampled_errors.loc[day - 1, "inflow"]
            )

            transfer = max(
                0,
                baseline_rates["transfer"] *
                assumptions["transfer_multiplier"] +
                sampled_errors.loc[day - 1, "transfer"]
            )

            discharge = max(
                0,
                baseline_rates["discharge"] *
                assumptions["discharge_multiplier"] +
                sampled_errors.loc[day - 1, "discharge"]
            )

            cbp_sim = max(
                0,
                cbp_sim + inflow - transfer
            )

            hhs_sim = max(
                0,
                hhs_sim + transfer - discharge
            )

            if day in FORECAST_HORIZONS:

                simulation_path_records.append({

                    "Scenario": scenario_name,

                    "Simulation": simulation_id,

                    "Horizon": day,

                    "Simulated_HHS_Care": hhs_sim,

                    "Simulated_CBP_Custody": cbp_sim,

                    "Simulated_Total_Population": (
                        hhs_sim + cbp_sim
                    ),

                    "Capacity_Breach": (
                        hhs_sim > CAPACITY_THRESHOLD
                        if CAPACITY_THRESHOLD is not None
                        else pd.NA
                    )

                })

simulation_paths = pd.DataFrame(
    simulation_path_records
)


# ------------------------------------------------------------
# CALCULATE SCENARIO UNCERTAINTY
# ------------------------------------------------------------

for (scenario_name, horizon), group in (
    simulation_paths.groupby(
        ["Scenario", "Horizon"]
    )
):

    uncertainty_records.append({

        "Scenario": scenario_name,

        "Horizon": horizon,

        "HHS_Mean": (
            group["Simulated_HHS_Care"].mean()
        ),

        "HHS_Lower_80": (
            group["Simulated_HHS_Care"].quantile(0.10)
        ),

        "HHS_Upper_80": (
            group["Simulated_HHS_Care"].quantile(0.90)
        ),

        "HHS_Lower_95": (
            group["Simulated_HHS_Care"].quantile(0.025)
        ),

        "HHS_Upper_95": (
            group["Simulated_HHS_Care"].quantile(0.975)
        ),

        "HHS_Uncertainty_Width_80": (
            group["Simulated_HHS_Care"].quantile(0.90) -
            group["Simulated_HHS_Care"].quantile(0.10)
        ),

        "HHS_Uncertainty_Width_95": (
            group["Simulated_HHS_Care"].quantile(0.975) -
            group["Simulated_HHS_Care"].quantile(0.025)
        ),

        "CBP_Mean": (
            group["Simulated_CBP_Custody"].mean()
        ),

        "CBP_Lower_95": (
            group["Simulated_CBP_Custody"].quantile(0.025)
        ),

        "CBP_Upper_95": (
            group["Simulated_CBP_Custody"].quantile(0.975)
        )

    })

scenario_uncertainty = pd.DataFrame(
    uncertainty_records
)

print("\nScenario uncertainty comparison:")

display(scenario_uncertainty)


# ------------------------------------------------------------
# CALCULATE CAPACITY BREACH PROBABILITY
# ------------------------------------------------------------

if CAPACITY_THRESHOLD is not None:

    capacity_breach_probability = (
        simulation_paths.groupby(
            ["Scenario", "Horizon"]
        )
        .agg(
            Capacity_Breach_Probability=(
                "Capacity_Breach", "mean"
            ),

            Simulation_Count=(
                "Capacity_Breach", "count"
            )
        )
        .reset_index()
    )

    capacity_breach_probability[
        "Capacity_Breach_Probability"
    ] *= 100

else:

    capacity_breach_probability = pd.DataFrame(
        columns=[
            "Scenario",
            "Horizon",
            "Capacity_Breach_Probability",
            "Simulation_Count"
        ]
    )

    print(
        "\nCapacity breach probability unavailable: "
        "verified operational capacity has not been supplied."
    )

print("\nCapacity breach probability:")

display(capacity_breach_probability)


# ------------------------------------------------------------
# COMPARE OPERATIONAL RESOURCE PRESSURE
# ------------------------------------------------------------

resource_records = []

for _, row in scenario_summary.iterrows():

    scenario = row["Scenario"]

    horizon = row["Horizon"]

    hhs_pressure = (
        row["Projected_HHS_Care"] /
        max(INITIAL_HHS_POPULATION, 1)
    ) * 100

    cbp_pressure = (
        row["Projected_CBP_Custody"] /
        max(INITIAL_CBP_POPULATION, 1)
    ) * 100

    resource_records.append({

        "Scenario": scenario,

        "Horizon": horizon,

        "HHS_Resource_Pressure_Index": hhs_pressure,

        "CBP_Resource_Pressure_Index": cbp_pressure,

        "Combined_Population": (
            row["Projected_Total_Population"]
        ),

        "Projected_Peak_Care_Load": (
            row["Projected_Peak_Care_Load"]
        ),

        "Capacity_Utilization": (
            row["Capacity_Utilization"]
        )

    })

resource_pressure = pd.DataFrame(
    resource_records
)

print("\nOperational resource pressure comparison:")

display(resource_pressure)


# ------------------------------------------------------------
# CREATE CONSOLIDATED SCENARIO COMPARISON TABLE
# ------------------------------------------------------------

scenario_comparison = scenario_summary.merge(
    scenario_uncertainty,
    on=["Scenario", "Horizon"],
    how="left"
)

if not capacity_breach_probability.empty:

    scenario_comparison = scenario_comparison.merge(
        capacity_breach_probability,
        on=["Scenario", "Horizon"],
        how="left"
    )

else:

    scenario_comparison[
        "Capacity_Breach_Probability"
    ] = np.nan

scenario_comparison = scenario_comparison.merge(
    resource_pressure,
    on=["Scenario", "Horizon"],
    how="left",
    suffixes=("", "_Resource")
)

print("\nCONSOLIDATED SCENARIO COMPARISON:")

display(scenario_comparison)


# ------------------------------------------------------------
# GENERATE OPERATIONAL RECOMMENDATIONS
# ------------------------------------------------------------

recommendations = []

for horizon in FORECAST_HORIZONS:

    horizon_data = scenario_comparison[
        scenario_comparison["Horizon"] == horizon
    ]

    if horizon_data.empty:
        continue

    low = horizon_data[
        horizon_data["Scenario"] == "Low_Demand"
    ]

    baseline = horizon_data[
        horizon_data["Scenario"] == "Baseline"
    ]

    high = horizon_data[
        horizon_data["Scenario"] == "High_Demand"
    ]

    if low.empty or baseline.empty or high.empty:
        continue

    low_hhs = low.iloc[0]["Projected_HHS_Care"]

    baseline_hhs = baseline.iloc[0]["Projected_HHS_Care"]

    high_hhs = high.iloc[0]["Projected_HHS_Care"]

    recommendations.append({

        "Horizon": horizon,

        "Low_Demand_HHS": low_hhs,

        "Baseline_HHS": baseline_hhs,

        "High_Demand_HHS": high_hhs,

        "Scenario_Range": high_hhs - low_hhs,

        "Operational_Recommendation": (
            "Prepare contingency staffing and placement plans "
            "for the high-demand scenario; monitor daily inflows, "
            "transfers and discharge rates."
            if high_hhs > baseline_hhs
            else
            "Monitor actual population and reassess staffing "
            "and placement requirements."
        )

    })

operational_recommendations = pd.DataFrame(
    recommendations
)

print("\nOPERATIONAL RECOMMENDATIONS:")

display(operational_recommendations)


# ------------------------------------------------------------
# VISUALIZATION: HHS CARE POPULATION SCENARIOS
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(14, 6))

for scenario_name, group in scenario_forecasts_df.groupby(
    "Scenario"
):

    ax.plot(
        group["Forecast_Date"],
        group["Projected_HHS_Care"],
        label=scenario_name,
        linewidth=2
    )

ax.set_title(
    "Projected HHS Care Population Under Alternative Scenarios"
)

ax.set_xlabel("Forecast Date")
ax.set_ylabel("Projected Children in HHS Care")

ax.legend()
ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    SCENARIO_DIR / "hhs_scenario_forecasts.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# VISUALIZATION: CBP CUSTODY POPULATION SCENARIOS
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(14, 6))

for scenario_name, group in scenario_forecasts_df.groupby(
    "Scenario"
):

    ax.plot(
        group["Forecast_Date"],
        group["Projected_CBP_Custody"],
        label=scenario_name,
        linewidth=2
    )

ax.set_title(
    "Projected CBP Custody Population Under Alternative Scenarios"
)

ax.set_xlabel("Forecast Date")
ax.set_ylabel("Projected Children in CBP Custody")

ax.legend()
ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    SCENARIO_DIR / "cbp_scenario_forecasts.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# VISUALIZATION: TOTAL POPULATION
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(14, 6))

for scenario_name, group in scenario_forecasts_df.groupby(
    "Scenario"
):

    ax.plot(
        group["Forecast_Date"],
        group["Projected_Total_Population"],
        label=scenario_name,
        linewidth=2
    )

ax.set_title(
    "Projected Combined HHS Care and CBP Custody Population"
)

ax.set_xlabel("Forecast Date")
ax.set_ylabel("Combined Population")

ax.legend()
ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    SCENARIO_DIR / "combined_population_scenarios.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# VISUALIZATION: SCENARIO UNCERTAINTY
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(13, 6))

for scenario_name, group in scenario_uncertainty.groupby(
    "Scenario"
):

    group = group.sort_values("Horizon")

    ax.plot(
        group["Horizon"],
        group["HHS_Mean"],
        marker="o",
        label=scenario_name
    )

    ax.fill_between(
        group["Horizon"],
        group["HHS_Lower_95"],
        group["HHS_Upper_95"],
        alpha=0.10
    )

ax.set_title(
    "Scenario-Specific HHS Population and 95% Uncertainty Ranges"
)

ax.set_xlabel("Forecast Horizon (Days)")
ax.set_ylabel("Projected HHS Care Population")

ax.set_xticks(FORECAST_HORIZONS)

ax.legend()
ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    SCENARIO_DIR / "scenario_uncertainty.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# VISUALIZATION: CAPACITY BREACH PROBABILITY
# ------------------------------------------------------------

if not capacity_breach_probability.empty:

    fig, ax = plt.subplots(figsize=(11, 6))

    for scenario_name, group in (
        capacity_breach_probability.groupby("Scenario")
    ):

        ax.plot(
            group["Horizon"],
            group["Capacity_Breach_Probability"],
            marker="o",
            label=scenario_name
        )

    ax.set_title(
        "Capacity Breach Probability Across Scenarios"
    )

    ax.set_xlabel("Forecast Horizon (Days)")
    ax.set_ylabel("Breach Probability (%)")

    ax.set_xticks(FORECAST_HORIZONS)

    ax.legend()
    ax.grid(alpha=0.25)

    plt.tight_layout()

    plt.savefig(
        SCENARIO_DIR / "capacity_breach_probability.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# VISUALIZATION: OPERATIONAL RESOURCE PRESSURE
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(12, 6))

for scenario_name, group in resource_pressure.groupby(
    "Scenario"
):

    ax.plot(
        group["Horizon"],
        group["HHS_Resource_Pressure_Index"],
        marker="o",
        label=scenario_name
    )

ax.axhline(
    100,
    linestyle="--",
    label="Current HHS Population Baseline"
)

ax.set_title(
    "Relative HHS Operational Resource Pressure"
)

ax.set_xlabel("Forecast Horizon (Days)")
ax.set_ylabel("Relative Resource Pressure (%)")

ax.set_xticks(FORECAST_HORIZONS)

ax.legend()
ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    SCENARIO_DIR / "operational_resource_pressure.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# SAVE ALL SCENARIO OUTPUTS
# ------------------------------------------------------------

output_tables = {

    "scenario_assumptions.csv":
        scenario_assumptions_df,

    "baseline_flow_rates.csv":
        pd.DataFrame([
            {
                "Flow": name,
                "Baseline_Daily_Rate": value
            }
            for name, value in baseline_rates.items()
        ]),

    "scenario_daily_forecasts.csv":
        scenario_forecasts_df,

    "scenario_forecast_summary.csv":
        scenario_summary,

    "hhs_scenario_comparison.csv":
        hhs_scenario_comparison,

    "cbp_scenario_comparison.csv":
        cbp_scenario_comparison,

    "peak_care_load.csv":
        peak_load_summary,

    "scenario_uncertainty.csv":
        scenario_uncertainty,

    "scenario_simulation_paths.csv":
        simulation_paths,

    "capacity_breach_probability.csv":
        capacity_breach_probability,

    "operational_resource_pressure.csv":
        resource_pressure,

    "scenario_comparison.csv":
        scenario_comparison,

    "operational_recommendations.csv":
        operational_recommendations

}

for filename, table in output_tables.items():

    table.to_csv(
        SCENARIO_DIR / filename,
        index=False
    )


# ------------------------------------------------------------
# DOCUMENT SCENARIO METHODOLOGY AND LIMITATIONS
# ------------------------------------------------------------

scenario_documentation = """
SCENARIO ANALYSIS AND OPERATIONAL PLANNING
==========================================

PROJECT:
Predictive Forecasting of Care Load & Placement Demand

SCENARIOS
---------

LOW-DEMAND SCENARIO
- Inflow multiplier: 0.80
- Transfer multiplier: 0.90
- Discharge multiplier: 1.10

BASELINE SCENARIO
- Inflow multiplier: 1.00
- Transfer multiplier: 1.00
- Discharge multiplier: 1.00

HIGH-DEMAND SCENARIO
- Inflow multiplier: 1.20
- Transfer multiplier: 1.10
- Discharge multiplier: 0.90

These multipliers are configurable planning assumptions.
They are not probabilities or official forecasts.

FLOW DEFINITIONS
----------------

CBP INFLOW:
Children apprehended and placed in CBP custody.

CBP TRANSFERS:
Children transferred out of CBP custody.

HHS INFLOW:
CBP transfers are treated as HHS inflows for the
simplified stock-flow model.

HHS DISCHARGES:
Children discharged from HHS care.

STOCK-FLOW EQUATIONS
--------------------

CBP Net Flow:
    CBP Inflow - CBP Transfers

HHS Net Flow:
    HHS Transfers - HHS Discharges

Projected CBP Population:
    Previous CBP Population + CBP Net Flow

Projected HHS Population:
    Previous HHS Population + HHS Net Flow

REPORTING FREQUENCY
-------------------

Observed flows are normalized by elapsed calendar days
to estimate approximate daily flow rates.

Missing reporting dates are not treated as zero-flow days.

The model assumes observed flow counts represent totals
accumulated during the intervals between reports.
This interpretation must be verified against source
documentation before operational deployment.

UNCERTAINTY
-----------

Historical flow residual bootstrap is used to generate
alternative possible future trajectories.

The 80% and 95% simulation ranges describe the
conditional uncertainty within each scenario.

They are not probabilities that a particular scenario
will occur.

CAPACITY RISK
-------------

Capacity utilization:
    Projected HHS Population / Verified Capacity * 100

Capacity breach probability:
    Percentage of simulated HHS populations exceeding
    the verified capacity threshold.

No capacity value is invented.

OPERATIONAL RESOURCE PRESSURE
-----------------------------

Relative HHS resource pressure:
    Projected HHS Population / Initial HHS Population * 100

Relative CBP resource pressure:
    Projected CBP Population / Initial CBP Population * 100

These are relative population indicators, not direct
staffing or facility-capacity measurements.

LIMITATIONS
-----------

1. Scenario multipliers are assumptions, not causal
   estimates or official policy forecasts.

2. Historical flow rates may change under policy,
   operational or external conditions.

3. Transfers out of CBP custody are treated as HHS
   inflows. Actual transfer timing and destinations
   should be validated.

4. Stock-flow reconciliation may be imperfect because
   of reporting timing, unobserved movements and
   differences in indicator definitions.

5. Daily rate normalization assumes interval flow
   observations represent accumulated counts between
   reporting dates.

6. The simulation uses historical flow residuals and
   may not capture unprecedented events.

7. Simulation outcomes do not establish the probability
   of the low, baseline or high-demand scenario.

8. Capacity breach analysis requires an authoritative
   operational capacity input.

9. Projected population is not a direct measure of
   staffing, bed availability or placement readiness.

10. CBP and HHS populations are presented separately.
    Combined population should not be interpreted as a
    unique count without verifying that the populations
    are mutually exclusive at the relevant timestamps.

11. Operational recommendations are planning guidance
    and require review by authorized operational teams.

12. Results should be updated as new verified data
    and operational assumptions become available.
"""

with open(
    SCENARIO_DIR / "scenario_methodology_and_limitations.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write(scenario_documentation)


# ------------------------------------------------------------
# FINAL EXECUTION SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SCENARIO ANALYSIS COMPLETED")
print("=" * 80)

print("\nHistorical observations:", len(df))

print(
    "Forecast start date:",
    latest_date
)

print(
    "Initial HHS population:",
    INITIAL_HHS_POPULATION
)

print(
    "Initial CBP population:",
    INITIAL_CBP_POPULATION
)

print(
    "Forecast horizons:",
    FORECAST_HORIZONS
)

print(
    "Scenarios:",
    list(SCENARIO_ASSUMPTIONS.keys())
)

print(
    "Bootstrap simulations per scenario:",
    N_SIMULATIONS
)

print(
    "Capacity threshold:",
    CAPACITY_THRESHOLD
)

print("\nScenario comparison:")

display(scenario_comparison)

print("\nOperational recommendations:")

display(operational_recommendations)

print("\nGenerated files:")

for file_path in sorted(SCENARIO_DIR.iterdir()):

    if file_path.is_file():
        print(" -", file_path.name)

print("\nOutput directory:")
print(SCENARIO_DIR)

print("\nAll scenario analysis outputs generated.")

display(
    Markdown(
        "**Scenario analysis and operational planning completed.**"
    )
)
```
